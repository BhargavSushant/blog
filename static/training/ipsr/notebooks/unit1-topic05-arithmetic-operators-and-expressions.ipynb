## 2. Operators and expressions

An **operator** is a symbol that does a calculation, such as `+`. The values it works on are **operands**.
An **expression** is a piece of code that Python can work out to a single value, for example `3 + 4 * 2`.
A statement like `total = 3 + 4 * 2` stores the value of the expression in a name.

## 3. The arithmetic operators

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | addition | `7 + 2` | `9` |
| `-` | subtraction | `7 - 2` | `5` |
| `*` | multiplication | `7 * 2` | `14` |
| `/` | true division (always a float) | `7 / 2` | `3.5` |
| `//` | floor division | `7 // 2` | `3` |
| `%` | remainder (modulo) | `7 % 2` | `1` |
| `**` | power | `7 ** 2` | `49` |

````python
a, b = 7, 2
print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)
print("a // b =", a // b)
print("a % b  =", a % b)
print("a ** b =", a ** b)
````

## 4. The two kinds of division

`/` always gives a `float`, even when the answer is whole. `//` throws away the fraction and gives the **floor** (the next whole number going down).

````python
print(10 / 2, type(10 / 2))        # 5.0, a float
print(10 // 2, type(10 // 2))      # 5, an int
print(7 // 2, 7 / 2)
print(7.0 // 2)                    # floor division of floats gives a float: 3.0
````

**Division identity.** For any numbers, `a == (a // b) * b + (a % b)`. The quotient and remainder always fit together this way.

````python
a, b = 17, 5
q, r = a // b, a % b
print(q, r, q * b + r == a)
print(divmod(17, 5))               # both at once: (quotient, remainder)
````

## 5. The remainder operator `%`

`a % b` is what is left after dividing `a` by `b`. It is very useful:

````python
n = 14
print("even" if n % 2 == 0 else "odd")          # even/odd test

print(1234 % 10)         # last digit
print(1234 // 10)        # everything except the last digit

hours_now, hours_later = 10, 5
print((hours_now + hours_later) % 12)           # clock arithmetic: 10 o'clock + 5 hours

total_seconds = 3725
hours, rest = divmod(total_seconds, 3600)
minutes, seconds = divmod(rest, 60)
print(hours, "h", minutes, "min", seconds, "s")
````

## 6. Power and roots

`**` raises to a power. A power of `0.5` is a square root, and a negative power gives a fraction.

````python
print(2 ** 10)
print(9 ** 0.5)         # square root
print(2 ** -1)          # 1 / 2
print(pow(2, 10), pow(2, 10, 1000))    # pow(a, b, m) is (a ** b) % m, done quickly
````

## 7. Precedence: which operator goes first?

Like in school maths (BODMAS / PEMDAS), Python does some operators before others. From **highest to lowest**:

| Level | Operators |
|---|---|
| 1 | `( )` brackets |
| 2 | `**` |
| 3 | `+x`, `-x` (unary plus and minus) |
| 4 | `*`, `/`, `//`, `%` |
| 5 | `+`, `-` (binary) |

Operators on the same level are worked out **left to right**, except `**`, which goes **right to left**.

````python
print(3 + 4 * 2)        # 4 * 2 first: 11
print((3 + 4) * 2)      # brackets first: 14
print(20 - 5 - 3)       # left to right: (20 - 5) - 3 = 12
print(100 / 10 * 2)     # left to right: (100 / 10) * 2 = 20.0
print(2 ** 3 ** 2)      # right to left: 2 ** (3 ** 2) = 2 ** 9 = 512
print((2 ** 3) ** 2)    # 64
````

**Tip:** when in doubt, add brackets. They cost nothing and help the reader.

### Trace an expression step by step
`2 + 3 * 4 ** 2 // 5 - 1`

````python
print(4 ** 2)                   # 16
print(3 * 16)                   # 48
print(48 // 5)                  # 9
print(2 + 9 - 1)                # 10
print(2 + 3 * 4 ** 2 // 5 - 1)  # the whole thing: 10
````

## 8. Mixed types

* `int` with `int`: the result is `int` (except `/`, which gives `float`).
* `int` with `float`: the result is `float`.
* `bool` counts as 0 or 1.
* Text can be added to text and repeated with `*`, but not mixed with numbers.

````python
print(3 + 4, type(3 + 4))
print(3 + 4.0, type(3 + 4.0))
print(True + True + 3)
print("ab" + "cd", "ab" * 3)
````

## 10. Edge cases

### 10.1 Floor division and remainder with negative numbers
`//` rounds **down** (towards minus infinity), not towards zero. So `-7 // 2` is `-4`, not `-3`.
The remainder takes the **sign of the right operand**, so that the division identity still holds.

````python
print(7 // 2, -7 // 2, 7 // -2, -7 // -2)
print(7 % 3, -7 % 3, 7 % -3, -7 % -3)
a, b = -7, 2
print((a // b) * b + (a % b) == a)     # the identity still holds
print(int(-7 / 2))                     # int() cuts towards zero: -3, different from -7 // 2
````

This is handy: `-1 % 5` is `4`, so `% n` always gives a number from `0` to `n - 1` when `n` is positive. Other languages such as C or Java would give `-1`.

### 10.2 Division by zero

````python
for expr in ["1 / 0", "1 // 0", "1 % 0", "0 ** -1", "0 / 5"]:
    try:
        print(expr, "=", eval(expr))
    except ZeroDivisionError as e:
        print(expr, "-> ZeroDivisionError:", e)
````

Check the divisor before you divide:

````python
def safe_divide(a, b):
    if b == 0:
        return None
    return a / b

print(safe_divide(10, 4), safe_divide(10, 0))
````

### 10.3 Floats are approximate

````python
print(0.1 + 0.2)
print(1.1 * 3)
print(0.1 * 3 == 0.3, math.isclose(0.1 * 3, 0.3))
print(1 / 3)
print(10 % 0.3)           # float remainder shows rounding error too
````

### 10.4 Unary minus and `**`
`**` binds tighter than a minus sign in front of it, so `-2 ** 2` means `-(2 ** 2)`.

````python
print(-2 ** 2)       # -4
print((-2) ** 2)     # 4
print(2 ** -2)       # 0.25
print(-2 ** 3, (-2) ** 3)
````

### 10.5 A negative number to a fractional power gives a complex number

````python
print((-8) ** (1 / 3))
try:
    print(math.sqrt(-1))
except ValueError as e:
    print("ValueError:", e)
````

### 10.6 Big integers and big floats
Integers never overflow. Floats do: they become `inf` or raise an error.

````python
print(2 ** 200)
print(len(str(3 ** 1000)), "digits in 3 ** 1000")
try:
    print(10.0 ** 400)
except OverflowError as e:
    print("OverflowError:", e)
print(1e308 * 10)
````

### 10.7 `/` always gives a float, even for huge ints
Dividing two very large ints with `/` converts to float, which can lose digits. Use `//` to keep every digit.

````python
n = 10 ** 30
print(n / 10)
print(n // 10)
print(int(n / 10) == n // 10)
````

### 10.8 Rounding surprises
`round` ties go to the even number, and floats may be a little off from what you typed.

````python
print(round(2.5), round(3.5), round(-2.5))
print(round(2.675, 2))
print(int(2.9999999999999999))      # the literal is stored as 3.0
````

### 10.9 Augmented operators and precedence
The whole right side is worked out first.

````python
x = 10
x *= 2 + 3          # x = x * (2 + 3), not (x * 2) + 3
print(x)
````

## 11. Quick summary

| Operator | Meaning | Remember |
|---|---|---|
| `+ - *` | add, subtract, multiply | int with int gives int |
| `/` | true division | always float; error if divisor is 0 |
| `//` | floor division | rounds **down**: `-7 // 2 == -4` |
| `%` | remainder | has the sign of the divisor: `-7 % 3 == 2` |
| `**` | power | right to left; `-2 ** 2 == -4` |
| Precedence | `( )`, `**`, unary `-`, `* / // %`, `+ -` | same level: left to right (except `**`) |
| Identity | `a == (a // b) * b + a % b` | always true |
| Helpers | `divmod`, `abs`, `round`, `pow`, `math.sqrt`, `math.floor` | `import math` first |
