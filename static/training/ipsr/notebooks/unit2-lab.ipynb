## 2.2 Programming with and without AI

*HTML lab: Unit 2 lab, lesson 1*

**Without AI** a programmer writes the rules. **With AI (machine learning)** the programmer supplies examples and the program finds the rules. Below the same task, spotting a spam message, is done both ways.

````python
import numpy as np
# Without AI: hand-written rule
def rule_spam(n_links, n_caps_words):
    return n_links >= 3 or n_caps_words >= 5

# With AI: learn the best threshold from labelled examples
X = np.array([[0,0],[1,1],[0,2],[4,1],[3,6],[5,7],[1,0],[2,5],[6,2],[0,1]])   # links, capital words
y = np.array([0,0,0,1,1,1,0,1,1,0])                                          # 1 = spam
score = X.sum(axis=1)
best = max(range(0, 14), key=lambda t: ((score >= t) == y).mean())
print("learned rule: spam if links + capital words >=", best,
      "| training accuracy:", ((score >= best) == y).mean())
test = [(0, 0), (4, 4), (1, 3)]
for a, b in test:
    print((a, b), "hand rule:", rule_spam(a, b), "| learned:", a + b >= best)
````

## 2.4 Turing Test

*HTML lab: Unit 2 lab, lesson 3*

Alan Turing (1950) proposed: if a human judge, chatting by text, cannot tell the machine from a person, the machine passes. The test checks *behaviour*, not understanding. A tiny pattern-matching chatbot (ELIZA style) shows how easy it is to look clever and how it fails.

````python
import re
rules = [
 (r"\b(hello|hi)\b",      "Hello! How are you today?"),
 (r"\bI am (.*)",         "Why do you say you are {0}?"),
 (r"\bmy (mother|father|friend)\b", "Tell me more about your {0}."),
 (r"\b(sad|happy|angry)\b", "Why do you feel {0}?"),
]
def bot(text):
    for pat, reply in rules:
        m = re.search(pat, text, re.I)
        if m: return reply.format(*m.groups())
    return "Please go on."
for line in ["Hi there", "I am tired", "My mother called", "What is 17 times 23?", "I feel sad"]:
    print("You:", line, "\nBot:", bot(line))
````

Notice the bot answers the arithmetic question with 'Please go on.' A judge would catch it in one question. That is the idea behind the test.

## 2.5 Environment constraints

*HTML lab: Unit 2 lab, lesson 4*

An agent works inside limits set by its environment. Typical constraints: limited sensors (partial view), noisy data, limited time to decide, limited memory and energy, and rules that actions must obey. A good agent design starts by writing these down.

````python
import random
random.seed(1)
true_temp = 30.0
def noisy_sensor():
    return true_temp + random.gauss(0, 1.5)
readings = [round(noisy_sensor(), 1) for _ in range(5)]
print("single readings:", readings)
print("average of 5 :", round(sum(readings)/5, 2), "(true value is", true_temp, ")")
````

## 2.6 Agents and environments

*HTML lab: Unit 2 lab, lesson 4*

An **agent** perceives its environment through **sensors** and acts on it through **actuators**. The loop is: percept in, action out, environment changes.

````python
class Thermostat:
    def act(self, percept):          # percept = current temperature
        return "heat ON" if percept < 22 else "heat OFF"
room = 18.0
agent = Thermostat()
for step in range(6):
    action = agent.act(room)
    room += 1.5 if action == "heat ON" else -0.5
    print(f"step {step}: temp={room:4.1f}  action={action}")
````

## 2.10 Example: VacBot with PEAS

*HTML lab: Unit 2 lab, lesson 6*

**VacBot:** P = rooms cleaned over time minus energy used; E = two squares A and B that may be dirty; A = Left, Right, Suck; S = location sensor, dirt sensor. Below is a full simulation.

````python
import random
def run(agent, steps=12, seed=3):
    random.seed(seed)
    dirt = {"A": random.random() < 0.7, "B": random.random() < 0.7}
    loc, score = "A", 0
    log = []
    for t in range(steps):
        action = agent(loc, dirt[loc])
        if action == "Suck":
            if dirt[loc]: score += 10
            dirt[loc] = False
        else:
            loc = "B" if action == "Right" else "A"
            score -= 1
        log.append((t, action, loc, dict(dirt), score))
    return log

def reflex_vacbot(loc, is_dirty):
    if is_dirty: return "Suck"
    return "Right" if loc == "A" else "Left"

for row in run(reflex_vacbot)[:8]:
    print(row)
print("final score:", run(reflex_vacbot)[-1][-1])
````

The score rewards sucking dirt (+10) and charges 1 for each move. That is a **performance measure** in action.
