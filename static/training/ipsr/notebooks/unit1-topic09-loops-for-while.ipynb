## 2. Why loops?

Computers are good at doing the same thing many times. Without a loop, printing the numbers 1 to 5 needs five `print` lines.
With a loop, one line is enough, and it works just as well for 5000 numbers.

Python has two loops:

* `for`: repeat **for each item** in a collection, or a fixed number of times
* `while`: repeat **as long as** a condition is true

## 3. The `for` loop

```
for variable in collection:
    body (indented)
```

Python takes the items one at a time, puts each in the variable, and runs the body.

````python
for fruit in ["apple", "banana", "mango"]:
    print("I like", fruit)
print("done")
````

The loop variable (`fruit`) is a normal name. Choose a meaningful one. After the loop it keeps the last value.

### Looping over a string
A string is a collection of characters.

````python
for ch in "Python":
    print(ch)
````

### Looping over `range`
`range(n)` gives `0, 1, ... n-1`. The stop value is not included. (More on `range` in the previous topic.)

````python
for i in range(5):
    print("step", i)

for i in range(1, 6):
    print(i, "squared is", i * i)

for i in range(10, 0, -3):
    print(i, end=" ")
print()
````

If you do not need the loop variable, call it `_` by habit.

````python
for _ in range(3):
    print("Hello!")
````

## 4. The accumulator pattern

Many loops build up one answer. Start with a **starting value** before the loop, and **update it** inside the loop.

* total: start at `0`, then `total = total + item`
* count: start at `0`, then add `1` when something matches
* product: start at `1`, then multiply
* biggest: start with the first item, then replace it when you see a bigger one

````python
marks = [72, 45, 88, 31, 90]

total = 0
for m in marks:
    total = total + m
print("total:", total)
print("average:", total / len(marks))

count = 0
for m in marks:
    if m >= 40:
        count += 1               # same as count = count + 1
print("passed:", count)

product = 1
for n in range(1, 6):
    product *= n
print("5 factorial:", product)

biggest = marks[0]
for m in marks:
    if m > biggest:
        biggest = m
print("highest:", biggest)
````

Building a **new list** is the same idea. Start with an empty list and `append`.

````python
squares = []
for n in range(1, 6):
    squares.append(n * n)
print(squares)

vowels = 0
for ch in "programming language":
    if ch in "aeiou":
        vowels += 1
print("vowels:", vowels)
````

## 5. The `while` loop

```
while condition:
    body
```

Python checks the condition **before every round**. If it is true, it runs the body and checks again. When it becomes false, the loop ends.
Use `while` when you do not know in advance how many rounds you need.

The body must **change something** so the condition can eventually become false.

````python
count = 1
while count <= 5:
    print(count)
    count += 1          # without this line the loop would never end
print("finished, count is now", count)
````

````python
# How many times can we halve 100 before it drops below 1?
x = 100
steps = 0
while x >= 1:
    x = x / 2
    steps += 1
print(steps, "halvings, x =", x)

# digits of a number
n = 90417
digits = 0
while n > 0:
    n = n // 10          # drop the last digit
    digits += 1
print("digits:", digits)
````

### When to use which
| Use `for` when | Use `while` when |
|---|---|
| you have a collection to go through | you repeat until something happens |
| you know the number of rounds | you do not know the number of rounds |

## 6. `break` and `continue`

* `break` leaves the loop **immediately**.
* `continue` skips the rest of this round and goes to the **next** round.

````python
# break: stop at the first number divisible by 7
for n in range(1, 50):
    if n % 7 == 0:
        print("first multiple of 7:", n)
        break

# continue: skip the even numbers
for n in range(1, 10):
    if n % 2 == 0:
        continue
    print(n, end=" ")
print()
````

### Loop `else`
A loop can have an `else` block. It runs **only if the loop finished normally**, without a `break`. This is perfect for "search, and tell me if not found".

````python
def check_prime(n):
    if n < 2:
        print(n, "is not prime")
        return
    for d in range(2, n):
        if n % d == 0:
            print(n, "is divisible by", d, "so it is not prime")
            break
    else:                        # no break happened: no divisor found
        print(n, "is prime")

for n in [7, 12, 13, 1]:
    check_prime(n)
````

## 7. Nested loops

A loop inside a loop. The inner loop runs completely for every round of the outer loop.

````python
for row in range(1, 4):
    for col in range(1, 4):
        print(row * col, end="\t")
    print()                      # new line after each row
````

````python
# a triangle of stars
for n in range(1, 6):
    print("*" * n)

# the same with an inner loop
for n in range(1, 4):
    line = ""
    for _ in range(n):
        line += "*"
    print(line)
````

`break` only leaves the **innermost** loop.

````python
for i in range(3):
    for j in range(3):
        if j == 2:
            break            # leaves only the j loop
        print(i, j)
````

## 8. `enumerate` and `zip`

### `enumerate`: item and position together
Beginners often write `for i in range(len(items))`. `enumerate` is simpler and safer.

````python
names = ["Asha", "Ravi", "Meena"]
for position, name in enumerate(names):
    print(position, name)

for rank, name in enumerate(names, start=1):      # count from 1
    print(f"{rank}. {name}")
````

### `zip`: walk through two lists side by side
`zip` pairs up the first items, then the second items, and so on. It stops when the **shorter** list ends.

````python
names = ["Asha", "Ravi", "Meena"]
marks = [80, 65, 92]
for name, mark in zip(names, marks):
    print(name, "scored", mark)

print(list(zip([1, 2, 3], "ab")))       # the 3 is dropped: shorter side wins
````

## 9. Edge cases and common mistakes

**1. Infinite loops.** If the condition never becomes false, the program never stops. In a notebook you must interrupt the kernel.
To study this safely, add an **iteration cap**: a counter that forces a `break`.

````python
# Bug: x never changes, so x < 5 stays true forever.
x = 0
rounds = 0
while x < 5:
    rounds += 1
    if rounds > 1000:          # safety cap for this demo
        print("stopped by the cap: this loop would run forever")
        break

# Fixed: the body moves x towards the exit.
x = 0
while x < 5:
    x += 1
print("fixed loop ended with x =", x)
````

A more sneaky version: a condition that can be skipped over. If `x` goes up by 3 and we test `x != 10`, then 10 is never hit.
Prefer `<` or `<=` over `!=` in loop conditions.

````python
x = 0
rounds = 0
while x != 10:
    x += 3
    rounds += 1
    if rounds > 20:
        print("cap reached, x =", x, "(we jumped over 10)")
        break

x = 0
while x < 10:
    x += 3
print("with <, loop ends at x =", x)
````

**2. Off-by-one errors.** The most common loop bug. Count carefully: `range(1, 5)` gives only 4 numbers, and the stop value is excluded.

````python
print("Want 1 to 5:")
print(list(range(1, 5)), "<- wrong, misses 5")
print(list(range(1, 6)), "<- right")

items = ["a", "b", "c"]
print(len(items), "items, last position is", len(items) - 1)
try:
    for i in range(len(items) + 1):     # one too many
        print(items[i])
except IndexError as e:
    print("IndexError:", e)
````

**3. Changing a list while looping over it.** When you remove items, the later items slide forward, and the loop skips one. This gives wrong answers with no error.

````python
numbers = [1, 2, 2, 3, 4, 4, 5]
for n in numbers:
    if n % 2 == 0:
        numbers.remove(n)        # dangerous
print(numbers)                   # still contains even numbers
````

The output is `[1, 2, 3, 4, 5]`: one `2` and one `4` survived. The safe ways are to loop over a **copy**, or to build a **new list**.

````python
numbers = [1, 2, 2, 3, 4, 4, 5]

# Way 1: build a new list of the items you want to keep
odds = []
for n in numbers:
    if n % 2 != 0:
        odds.append(n)
print(odds)

# Way 2: loop over a copy, change the original
for n in numbers[:]:
    if n % 2 == 0:
        numbers.remove(n)
print(numbers)
````

**4. A loop over an empty collection does nothing**, and a `while` with a false start also runs **zero** times. Both are correct, but check your code still behaves sensibly.

````python
for x in []:
    print("never printed")

n = 10
while n < 5:
    print("never printed")

total = 0
for x in []:
    total += x
print("sum of nothing is", total)
````

**5. The loop variable stays after the loop, and is undefined if the loop never ran.**

````python
for i in range(3):
    pass
print("after the loop, i is", i)

for j in range(0):
    pass
try:
    print(j)
except NameError as e:
    print("NameError:", e)
````

**6. Changing the loop variable inside a `for` loop does nothing to the loop.** The next round takes the next item from the collection anyway.

````python
for i in range(3):
    print("before", i, end=" ")
    i = i + 100
    print("after", i)
````

**7. `continue` in a `while` loop can cause an infinite loop** if the update line is skipped. Put the update **before** `continue`.

````python
i = 0
while i < 6:
    i += 1                 # update first
    if i % 2 == 0:
        continue
    print(i, end=" ")
print()
````

**8. Float steps in `while` loops accumulate small errors.** Adding `0.1` ten times does not give exactly `1.0`, so `while x != 1.0` may never end. Count whole numbers and divide.

````python
x = 0.0
for _ in range(10):
    x += 0.1
print(x, x == 1.0)

for k in range(11):
    print(k / 10, end=" ")
print()
````

## 10. Quick summary

| Idea | Remember |
|---|---|
| `for x in collection:` | one round per item. Works on lists, strings, ranges |
| `range(a, b, s)` | `a` up to but not including `b`, step `s` |
| `while condition:` | repeat while true. The body must change something |
| Accumulator | set a start value before the loop, update inside |
| `break` | leave the loop now |
| `continue` | jump to the next round |
| loop `else` | runs only if there was **no** `break` |
| Nested loops | inner loop runs fully for each outer round. `break` leaves the inner one only |
| `enumerate(x, start=1)` | position and item together |
| `zip(a, b)` | pairs items, stops at the shorter list |
| Dangers | infinite loops, off-by-one, editing a list while looping over it |
