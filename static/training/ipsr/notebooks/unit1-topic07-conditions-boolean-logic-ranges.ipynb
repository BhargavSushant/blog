## 2. Booleans

A **boolean** (type `bool`) has only two values: `True` and `False`. Note the capital first letter.
George Boole, a mathematician, studied this kind of logic, which is why we call it boolean.
Programs use booleans to make decisions.

````python
is_raining = True
has_umbrella = False
print(is_raining, has_umbrella)
print(type(is_raining))
````

Most booleans come from **comparisons**. A comparison asks a yes or no question about two values.

## 3. Comparison operators

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | equal to | `5 == 5` | `True` |
| `!=` | not equal to | `5 != 3` | `True` |
| `<` | less than | `3 < 5` | `True` |
| `<=` | less than or equal to | `5 <= 5` | `True` |
| `>` | greater than | `3 > 5` | `False` |
| `>=` | greater than or equal to | `5 >= 6` | `False` |

````python
age = 17
print(age == 18)
print(age != 18)
print(age < 18)
print(age >= 18)
````

Comparisons work on text too. Text is compared letter by letter, in dictionary order, but **capital letters come before small letters**.

````python
print("apple" < "banana")
print("apple" == "Apple")      # case matters
print("Zebra" < "apple")        # capital Z comes before small a
print("10" < "9")               # text, not numbers: '1' comes before '9'
print(10 < 9)
````

### Chained comparisons
In maths we write `0 < x < 10`. Python allows the same thing. It means `0 < x` **and** `x < 10`.

````python
x = 7
print(0 < x < 10)
print(0 < x < 5)
print(1 <= x <= 7)       # both ends included
print(1 < x == 7 > 3)    # legal, but hard to read. Avoid this style.
````

## 4. Logical operators: `and`, `or`, `not`

| Operator | Result is `True` when |
|---|---|
| `a and b` | **both** are true |
| `a or b` | **at least one** is true |
| `not a` | `a` is false |

Here is the **truth table**, showing every case.

````python
print("A      B      A and B   A or B")
for A in (True, False):
    for B in (True, False):
        print(f"{str(A):6} {str(B):6} {str(A and B):9} {A or B}")

print("not True  =", not True)
print("not False =", not False)
````

A real example: a person can vote if they are at least 18 **and** they are a citizen.

````python
age = 20
is_citizen = True
can_vote = age >= 18 and is_citizen
print(can_vote)

day = "Sunday"
is_holiday = day == "Saturday" or day == "Sunday"
print(is_holiday)

print(not (age < 18))
````

### Precedence: which operator goes first?
From strongest to weakest: **comparisons**, then `not`, then `and`, then `or`.
So `and` binds tighter than `or`, like multiplication before addition. When in doubt, add brackets. They cost nothing.

````python
print(True or False and False)       # and first:  True or (False and False) -> True
print((True or False) and False)     # brackets change it -> False

a, b, c = 5, 10, 15
print(a < b and b < c or c < a)
print(not a > b)                     # not (a > b)
````

### Short-circuit evaluation
Python stops as soon as the answer is known.

* `A and B`: if `A` is false, the answer is false, so `B` is not even looked at.
* `A or B`: if `A` is true, the answer is true, so `B` is not looked at.

This is useful for guarding against errors.

````python
items = []
# len(items) > 0 is False, so the second part (which would fail) never runs
print(len(items) > 0 and items[0] == "pen")

y = 0
print(y != 0 and 10 / y > 1)         # no ZeroDivisionError
````

## 5. Truthiness: values that act like `True` or `False`

Python lets you use **any** value as a condition. `bool(value)` tells you what Python thinks of it.

**Falsy** values (count as `False`): `0`, `0.0`, `""` (empty text), `[]` (empty list), `{}`, `()`, `set()`, `None`, and `False` itself.
**Everything else is truthy.**

````python
for value in [0, 0.0, "", [], {}, None, False, 1, -3, 0.1, "a", " ", "0", [0], [[]]]:
    print(f"{value!r:8} -> {bool(value)}")
````

Notice: `"0"` and `" "` (a space) are **not** empty text, and `[0]` is a list with one item, so all three are truthy.

This gives short, readable code.

````python
name = ""
if not name:
    print("Please type your name")

basket = ["pen", "book"]
if basket:
    print("Basket has", len(basket), "items")
````

### `and` and `or` return a value, not only `True` or `False`
* `A or B` gives `A` if `A` is truthy, otherwise `B`.
* `A and B` gives `A` if `A` is falsy, otherwise `B`.

A common use is giving a **default value**.

````python
typed = ""
print(typed or "Guest")          # '' is falsy, so we get "Guest"
typed = "Asha"
print(typed or "Guest")

print(0 and 5, "|", 3 and 5, "|", 0 or 5, "|", 3 or 5)
````

## 6. Ranges

`range()` makes a sequence of whole numbers. It is used mostly in `for` loops.

* `range(stop)` starts at 0 and goes up to **but not including** `stop`
* `range(start, stop)` starts at `start`
* `range(start, stop, step)` moves by `step` each time

The rule to remember: **the stop value is exclusive**. `range(5)` gives 0, 1, 2, 3, 4. A `range` is lazy, so we wrap it in `list()` to see the numbers.

````python
print(list(range(5)))
print(list(range(2, 8)))
print(list(range(0, 20, 5)))
print(list(range(1, 11, 2)))        # odd numbers
````

### Negative step (counting down)
With a negative step the numbers go down. Then `start` must be **bigger** than `stop`.

````python
print(list(range(5, 0, -1)))        # 5 down to 1 (0 is excluded)
print(list(range(10, 0, -3)))
print(list(range(5, -1, -1)))       # include 0 by stopping at -1
````

### Empty ranges
If the numbers cannot move from `start` towards `stop` in the direction of `step`, the range is simply empty. This is not an error.

````python
print(list(range(0)))
print(list(range(5, 2)))            # going up, but 2 is below 5
print(list(range(2, 5, -1)))        # going down, but 5 is above 2
print(len(range(5, 2)))
````

### Useful facts about `range`
* `len(range(a, b))` is `b - a` when `a < b` (the count of numbers).
* `x in range(...)` tests membership without building a list.
* `step` cannot be `0`.

````python
r = range(10, 50, 10)
print(len(r), r[0], r[-1])
print(30 in r, 35 in r, 50 in r)         # 50 is the stop value, so not included
print(5 in range(1, 10))                 # a neat way to test 'between'
print(5.0 in range(1, 10))               # equal to 5, so True
print(5.5 in range(1, 10))
print(range(3))                          # a range is not a list
print(type(range(3)))
````

## 7. `==` versus `is`

* `==` asks: **do the two values look the same?**
* `is` asks: **are they the very same object in memory?**

For comparing numbers and text you want `==`. Use `is` only for `None` (write `x is None`) and for `True` / `False` checks on objects.

````python
a = [1, 2, 3]
b = [1, 2, 3]
c = a
print(a == b)      # same contents
print(a is b)      # two separate lists
print(a is c)      # c is another name for the same list

value = None
print(value is None)
print(value is not None)
````

## 8. Comparing decimal numbers

Computers store decimals in binary, and many decimals cannot be stored exactly. So `==` on floats can surprise you.
Use `math.isclose(a, b)`, which says whether two numbers are equal within a tiny tolerance.

````python
import math

print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
print(math.isclose(0.1 + 0.2, 0.3))

print(1.1 * 3)
print(1.1 * 3 == 3.3, math.isclose(1.1 * 3, 3.3))
````

## 9. Edge cases

**1. A chain checks each neighbouring pair.** `1 < 3 < 2` means `1 < 3 and 3 < 2`. The first part is true, the second is false, so the whole chain is false.
Chains are also legal with `==`, such as `a == b == c`, but keep them simple: use them mainly for ranges like `a <= x <= b`.

````python
print(1 < 2 < 3)
print(1 < 3 < 2)         # 1 < 3 is True but 3 < 2 is False
print(5 > 3 > 4)         # 3 > 4 is False
````

**2. `x == 1 or 2` is a classic mistake.** It reads as `(x == 1) or 2`, and `2` is always truthy, so the whole thing is always true.

````python
x = 7
print(x == 1 or 2)             # wrong idea, prints 2 (truthy)
print(bool(x == 1 or 2))       # True, even though x is 7!

print(x == 1 or x == 2)        # correct
print(x in (1, 2))             # also correct and shorter
````

**3. `True` and `False` are numbers too.** `True` is `1` and `False` is `0`. So you can add them, which is a nice way to count how many conditions hold.

````python
print(True + True, True * 5, False + 3)
print(True == 1, False == 0)

marks = [35, 80, 42, 90, 20]
print("students who passed:", sum(m >= 40 for m in marks))
````

**4. Comparing different types.** `==` between a number and text is simply `False`. But `<` between a number and text is a `TypeError`.

````python
print(5 == "5")
try:
    print(5 < "6")
except TypeError as e:
    print("TypeError:", e)
````

**5. `range` stop is exclusive, and `float` is not allowed.**

````python
print(list(range(1, 5)))           # 1, 2, 3, 4  (not 5)
try:
    range(0, 1, 0.1)
except TypeError as e:
    print("TypeError:", e)
try:
    range(0, 10, 0)
except ValueError as e:
    print("ValueError:", e)
````

**6. `is` with numbers and text is unreliable.** Python may or may not reuse objects, so the answer can change. Never use `is` to compare values.

````python
x = 1000
y = int("1000")
print(x == y)       # always the right question
# print(x is y)     # may be True or False. Do not rely on it.
````

**7. `NaN` is not equal to anything, even itself.**

````python
nan = float("nan")
print(nan == nan)
print(math.isnan(nan))
````

## 10. Quick summary

| Idea | Remember |
|---|---|
| `bool` | only `True` and `False`, capital first letter |
| Comparison | `==  !=  <  <=  >  >=` give a bool. `=` is assignment, not comparison |
| Chained | `0 < x < 10` means `0 < x and x < 10` |
| Logical | `not` first, then `and`, then `or`. Use brackets when unsure |
| Short-circuit | `and` / `or` stop as soon as the answer is known |
| Falsy values | `0`, `0.0`, `""`, `[]`, `{}`, `()`, `None`, `False`. All else truthy |
| `range(a, b, s)` | from `a` up to but **not including** `b`, step `s`. Can be empty |
| Countdown | negative step and `start > stop`: `range(5, 0, -1)` |
| `==` vs `is` | `==` same value, `is` same object. Use `is` only with `None` |
| Floats | use `math.isclose(a, b)`, not `==` |
