## 2. What is a variable?

A **variable** is a name for a value, so that you can use it again later. You create one with the **assignment** operator `=`.
Read `x = 5` as "let the name `x` point to the value 5".

(In maths, `x = 5` states a fact. In Python, it is an **action**: make `x` refer to 5.)

````python
length = 12
width = 5
area = length * width
print(area)
````

Python works out the **right side first**, then attaches the name on the left. So this is legal, and common:

````python
score = 10
score = score + 5      # right side: 10 + 5 = 15, then the name score points to 15
print(score)
````

You do not declare a type. A name can later point to a value of another type (this is **dynamic typing**).
Using a name that has no value yet gives a `NameError`.

````python
try:
    print(undefined_name)
except NameError as e:
    print("NameError:", e)
````

## 3. Naming rules and style (quick reminder)

* letters, digits and `_` only; cannot start with a digit; cannot be a keyword
* case matters: `age` and `Age` are different
* style: use **snake_case** for variables, for example `total_marks`
* choose meaningful names: `area` is better than `a`
* constants (values that should never change) are written in CAPITALS by convention, for example `GST_RATE = 0.18`. Python does **not** stop you from changing them.

## 4. Ways to assign

````python
# multiple assignment: several names, several values, in one line
a, b, c = 1, 2, 3
print(a, b, c)

# chained assignment: one value, several names
x = y = z = 0
print(x, y, z)

# swapping, without a temporary variable
a, b = b, a
print(a, b)
````

The right side `b, a` is built first as a pair, then unpacked into `a, b`. That is why the swap works.

### Unpacking
You can split a sequence into names. A name with a star `*` collects "the rest".

````python
first, second = [10, 20]
print(first, second)

head, *tail = [1, 2, 3, 4]
print(head, tail)

x, _, z = (7, "ignore me", 9)     # underscore is a name that means "I do not need this"
print(x, z)
````

### Augmented assignment
A short form for "update this variable using its own value".

````python
n = 10
n += 5      # n = n + 5
n -= 3      # n = n - 3
n *= 2      # n = n * 2
n //= 4     # n = n // 4
n **= 2     # n = n ** 2
print(n)
````

## 5. Names point to objects: `id()` and `is`

Every value in memory is an **object**. `id(x)` gives a number that identifies the object. `a is b` is `True` when both names point to the **same object**.
`a == b` is different: it asks whether the **values** are equal.

````python
a = [1, 2, 3]
b = a            # b points to the SAME list, no copy is made
c = [1, 2, 3]    # a new list with equal content

print(a == b, a is b)    # equal, and same object
print(a == c, a is c)    # equal, but different objects
print(id(a) == id(b), id(a) == id(c))
````

Think of names as **sticky labels** on objects. Two labels can stick on the same object.

## 6. Mutable and immutable objects

* **Immutable** objects cannot be changed after they are created: `int`, `float`, `bool`, `str`, `tuple`, `None`.
* **Mutable** objects can be changed in place: `list`, `dict`, `set`.

For an immutable object, an "update" really creates a **new object** and moves the name to it. The old object is untouched.

````python
n = 10
old_id = id(n)
n = n + 1                # a new int object 11; the name n moves to it
print("same object after n = n + 1?", id(n) == old_id)
````

````python
s = "hello"
try:
    s[0] = "J"                 # try to change one letter in place
except TypeError as e:
    print("TypeError:", e)

s2 = "J" + s[1:]               # build a NEW string instead
print(s, s2)
````

````python
t = (1, 2, 3)
try:
    t[0] = 99
except TypeError as e:
    print("TypeError:", e)
````

A `list` can be changed in place. The object stays the same, and its `id` does not change.

````python
marks = [50, 60, 70]
old_id = id(marks)
marks[0] = 55            # change in place
marks.append(80)         # change in place
print(marks)
print("same object after changes?", id(marks) == old_id)
````

### Why does this matter?
When two names share one **mutable** object, a change through one name shows up through the other.
With **immutable** objects, sharing is safe, because nobody can change the object.

````python
# immutable: safe
p = 5
q = p
q = q + 1
print(p, q)          # p is unchanged

# mutable: shared
a = [1, 2]
b = a
b.append(3)
print(a, b)          # a changed too!

# make a real copy when you need an independent list
c = a.copy()         # or list(a), or a[:]
c.append(4)
print(a, c)
````

### Passing to a function works the same way
A function receives a reference to the **same object**. If the object is mutable and the function changes it, the caller sees the change.

````python
def add_bonus(numbers):
    numbers.append(100)      # changes the caller's list

def add_one(n):
    n = n + 1                # only moves the local name n to a new int
    return n

data = [1, 2]
add_bonus(data)
print(data)

k = 5
add_one(k)
print(k)                    # still 5
````

## 7. Edge cases

### 7.1 Chained assignment with a mutable object
`a = b = []` makes **one** list with two names, not two lists.

````python
a = b = []
a.append(1)
print(a, b)

a = []
b = []          # two separate lists
a.append(1)
print(a, b)
````

### 7.2 A list of lists made with `*`
`[[0] * 3] * 2` repeats the **same** inner list twice.

````python
grid = [[0] * 3] * 2
grid[0][0] = 9
print(grid)              # both rows changed

grid = [[0] * 3 for _ in range(2)]      # a new inner list for each row
grid[0][0] = 9
print(grid)
````

### 7.3 `+=` is not always the same as `x = x + y`
For a list, `+=` changes the list **in place**, while `x = x + y` builds a **new** list.

````python
a = [1, 2]
b = a
a += [3]            # in place: b sees it
print(a, b, a is b)

a = [1, 2]
b = a
a = a + [3]         # new list: b still sees the old one
print(a, b, a is b)
````

### 7.4 An immutable tuple can hold a mutable list
The tuple cannot change **which objects it holds**, but a list inside it can still change.

````python
t = (1, [2, 3])
t[1].append(4)
print(t)

try:
    t[1] += [5]       # does the append, THEN fails on the assignment back to the tuple
except TypeError as e:
    print("TypeError:", e)
print(t)              # the list changed anyway
````

### 7.5 Small integers and `is`
Python keeps ready-made objects for small integers (about -5 to 256), so `is` may seem to work for them.
This is an implementation detail. **Never use `is` to compare numbers or text. Use `==`.**
`is` is for `None`, `True` and `False`, for example `x is None`.

````python
a = 256
b = 256
print(a is b)

x = int("100000")
y = int("100000")
print(x == y, x is y)      # equal values, but separate objects
````

### 7.6 Shadowing a built-in name
Naming a variable `list`, `str`, `sum` or `max` hides the built-in function.

````python
numbers = [4, 9, 2]
max = 100                       # shadows the built-in max
try:
    print(max(numbers))
except TypeError as e:
    print("TypeError:", e)
del max                         # the built-in is visible again
print(max(numbers))
````

### 7.7 A mutable default argument
The default value of a function parameter is created **once**, when the function is defined. A default list is then shared by all calls.

````python
def add_item(item, box=[]):
    box.append(item)
    return box

print(add_item("a"))
print(add_item("b"))       # surprise: 'a' is still there

def add_item_safe(item, box=None):
    if box is None:
        box = []           # a fresh list on every call
    box.append(item)
    return box

print(add_item_safe("a"))
print(add_item_safe("b"))
````

### 7.8 Deleting a name does not always delete the object
`del x` removes the **name**. The object lives on while another name still points to it.

````python
a = [1, 2, 3]
b = a
del a
print(b)
try:
    print(a)
except NameError as e:
    print("NameError:", e)
````

## 8. Quick summary

| Idea | Remember |
|---|---|
| Variable | a name that points to an object |
| `=` | evaluate the right side, then bind the name on the left |
| Multiple / chained / swap | `a, b = 1, 2` ; `a = b = 0` ; `a, b = b, a` |
| Augmented | `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=` |
| `==` vs `is` | equal value vs same object |
| Immutable | `int`, `float`, `bool`, `str`, `tuple`, `None`: "changing" makes a new object |
| Mutable | `list`, `dict`, `set`: change in place, shared names see the change |
| Copy a list | `b = a.copy()` or `list(a)` |
| Traps | `a = b = []`, `[[0]*3]*2`, mutable default arguments, shadowing built-ins |
