## 2. Statements and `print`

A Python program is a list of **statements**. Python runs them from top to bottom, one at a time.
Each statement normally sits on its own line. There is no `;` or `{ }` at the end of a line, unlike C.

The `print()` function shows a value on the screen.

````python
print("Hello, IPSR!")
print(2 + 3)
print("2 + 3 =", 2 + 3)      # print can take several values, separated by commas
````

`print` puts a space between values and a new line at the end. You can change both with `sep` and `end`.

````python
print("a", "b", "c", sep="-")
print("no newline here", end=" | ")
print("same line")
````

## 3. Indentation is part of the syntax

Many languages use `{ }` to group statements. Python uses **indentation** (the spaces at the start of a line).
A line that ends with a colon `:` starts a **block**, and the next lines must be indented by the same amount.
The usual rule is **4 spaces** per level.

````python
marks = 72
if marks >= 40:
    print("pass")          # this line belongs to the if block
    print("well done")     # so does this one
print("this line always runs")   # no indent: outside the block
````

Change `marks` to `30` and run again. Notice which lines still run.

### Edge case: wrong or mixed indentation
Python reports an error instead of guessing what you meant. We use `compile()` below so the notebook can show
the error message without stopping.

````python
def show_error(source):
    """Try to compile a piece of code and print the error Python gives."""
    try:
        compile(source, "<student code>", "exec")
        print("No syntax error")
    except SyntaxError as e:          # IndentationError is a kind of SyntaxError
        print(type(e).__name__ + ":", e.msg, "(line", e.lineno, ")")

show_error("if True:\nprint('x')")                 # block not indented
show_error("if True:\n    print('a')\n      print('b')")   # unexpected extra indent
show_error("if True:\n    print('a')\n  print('b')")        # indent does not match any outer level
show_error("    print('indent at top level')")      # indent where none is allowed
````

**Rule of thumb:** never mix tabs and spaces in the same file. Set your editor to insert 4 spaces when you press Tab.

## 4. Identifiers, keywords and case

An **identifier** is a name you choose, for a variable, function and so on. The rules:

* letters, digits and underscore `_` only
* cannot start with a digit
* cannot be a **keyword** (a word Python reserves, such as `if`, `for`, `class`)
* **case matters**: `total`, `Total` and `TOTAL` are three different names

````python
import keyword
print(len(keyword.kwlist), "keywords:")
print(keyword.kwlist)
````

````python
total = 10
Total = 20
print(total, Total, total == Total)

student_name = "Asha"     # good: lower case with underscores (snake_case)
x1 = 5                    # fine: digits are allowed after the first character
_hidden = 1               # fine: leading underscore is allowed (it signals 'internal use')
````

### Edge cases: names Python rejects

````python
for bad in ["1st_place = 1", "my-name = 1", "class = 1", "my name = 1", "total$ = 1"]:
    print(f"{bad:18}", end=" -> ")
    show_error(bad)
````

Two more traps:

* `True`, `False` and `None` are keywords with a capital first letter. `true` is just an ordinary (undefined) name.
* Python lets you reuse the name of a built-in function such as `print` or `sum`. It does not complain, but it breaks the built-in. This is called **shadowing**.

````python
try:
    print(true)
except NameError as e:
    print("NameError:", e)

list_of_numbers = [3, 4, 5]
sum = 100                     # oops: we replaced the built-in sum
try:
    print(sum(list_of_numbers))
except TypeError as e:
    print("TypeError:", e)

del sum                       # remove our variable; the built-in is visible again
print(sum(list_of_numbers))
````

## 5. Long statements and several statements on one line

A statement that is too long can be continued in two ways:

1. **Inside brackets** `( )`, `[ ]`, `{ }`: just press Enter. This is the preferred way.
2. With a backslash `\` as the very last character on the line.

You may put several statements on one line with `;`, but this is rarely good style.

````python
total = (10 +
         20 +
         30)
print(total)

also_total = 10 + \
             20 + \
             30
print(also_total)

a = 1; b = 2; print(a + b)      # legal, but hard to read
````

**Edge case:** a space after the backslash breaks the continuation, so the line must end exactly at the `\`.

````python
show_error("x = 1 + \\ \n    2")      # there is a space after the backslash
````

## 6. Comments

A **comment** starts with `#` and runs to the end of the line. Python ignores it completely. Comments are for people.

````python
# This whole line is a comment
price = 250          # a comment can also follow code on the same line
quantity = 4
print(price * quantity)   # print(price + quantity)  <- code that is 'commented out' never runs
````

### What makes a good comment
* explain **why**, not **what**: `# discount applies only to orders above 1000` is helpful, `# multiply price by quantity` is noise
* keep it true: when you change the code, change the comment
* use comments to switch a line off while testing ("commenting out"). In Jupyter and most editors, **Ctrl + /** toggles comments on the selected lines.

### Multi-line comments? There are none
Python has no block comment. Put a `#` at the start of every line.

### Docstrings are not comments
A string in triple quotes placed first inside a function, class or file is a **docstring**. Python keeps it, and `help()` shows it.

````python
def area_of_circle(r):
    """Return the area of a circle of radius r."""
    return 3.14159 * r * r

help(area_of_circle)
print(area_of_circle.__doc__)
````

**Edge case:** a triple-quoted string that is *not* in the docstring position is only a plain string value that Python builds and then throws away. It looks like a comment but is not one. Use `#` for comments.

````python
x = 5
"""I look like a comment, but I am a string
that Python created and ignored."""
print(x)
print("# this is inside a string, so it is NOT a comment")
````

## 7. Quick summary

| Idea | Rule |
|---|---|
| End of statement | end of line (no `;` needed) |
| Blocks | colon `:` then an indented block (4 spaces) |
| Names | letters, digits, `_`; not starting with a digit; not a keyword; case-sensitive |
| Long lines | continue inside brackets, or end the line with `\\` |
| Comments | `#` to end of line; no block comments |
| Docstring | triple-quoted string first in a function, class or file; not a comment |
