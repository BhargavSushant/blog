## 2. Creating strings

A **string** (`str`) is a sequence of characters. Write it inside single quotes, double quotes or triple quotes.
Single and double quotes work the same. Triple quotes allow several lines.

````python
a = 'single quotes'
b = "double quotes"
c = """triple quotes
can run over
several lines"""
print(a)
print(b)
print(c)
print(type(a))
````

Use the other kind of quote when your text contains a quote.

````python
print("It's easy")
print('He said "hello"')
print("")            # an empty string is allowed
print(len(""))
````

### Escape sequences
A backslash `\` starts an **escape sequence**: a special character written with two symbols.

| Escape | Meaning |
|---|---|
| `\n` | new line |
| `\t` | tab |
| `\\` | one backslash |
| `\'` and `\"` | a quote inside the same kind of quote |

````python
print("line one\nline two")
print("name\tmarks")
print("C:\\new_folder")
print('It\'s fine')
````

### Raw strings
Put `r` before the quote and Python keeps every backslash as it is. This is handy for Windows paths.

````python
path = r"C:\new_folder\table.txt"
print(path)
print(len(path))
print("C:\new_folder\table.txt")       # the \n and \t are treated as escapes here
````

**Edge case:** a raw string cannot end with a single backslash, because the backslash would escape the closing quote.

````python
try:
    compile('r"abc\\"', "<test>", "exec")
except SyntaxError as e:
    print("SyntaxError:", e.msg)

print(r"abc" + "\\")      # workaround: join a normal string with the backslash
````

## 3. Indexing and slicing

Each character has a position (an **index**). The first index is `0`. Negative indexes count from the end: `-1` is the last character.

```
 s =  P  y  t  h  o  n
index 0  1  2  3  4  5
neg  -6 -5 -4 -3 -2 -1
```

````python
s = "Python"
print(s[0], s[1], s[5])
print(s[-1], s[-2], s[-6])
print(len(s))
````

A **slice** `s[start:stop]` gives the characters from `start` up to, but **not including**, `stop`.
You can leave out either number. A third number is the step.

````python
s = "Python"
print(s[0:2])       # Py
print(s[2:])        # thon
print(s[:3])        # Pyt
print(s[-3:])       # hon
print(s[::2])       # Pto  (every second character)
print(s[::-1])      # nohtyP  (reversed)
````

### Edge case: out-of-range index and slice
An **index** outside the string is an error. A **slice** outside the string is quietly trimmed.

````python
s = "Python"
try:
    print(s[10])
except IndexError as e:
    print("IndexError:", e)

print(repr(s[2:100]))      # stops at the end
print(repr(s[10:20]))      # nothing there: empty string, no error
print(repr(s[4:2]))        # start is after stop: empty string
````

## 4. Strings are immutable

**Immutable** means you cannot change a string in place. You can only make a new string.

````python
s = "Python"
try:
    s[0] = "J"
except TypeError as e:
    print("TypeError:", e)

s = "J" + s[1:]        # build a new string and give it the old name
print(s)

t = "abc"
print(t.upper(), t)    # upper() returns a new string, t is unchanged
````

**Remember:** string methods never change the original. Always use the returned value.

## 5. Common methods

### Changing case and cleaning spaces

````python
s = "  Hello, World  "
print(s.upper())
print(s.lower())
print("hello world".title())
print(repr(s.strip()))       # remove spaces at both ends
print(repr(s.lstrip()))      # left only
print(repr(s.rstrip()))      # right only
print("xxhixx".strip("x"))   # strip other characters
````

### Splitting and joining
`split` cuts a string into a list. `join` does the opposite: the string before `.join` goes **between** the items.

````python
line = "apple,banana,cherry"
fruits = line.split(",")
print(fruits)

print("one  two   three".split())      # no argument: split on any white space
print("a-b-c".split("-", 1))           # at most 1 split

print(" + ".join(fruits))
print("".join(["a", "b", "c"]))
````

`join` needs a list of **strings**. Numbers must be converted first.

````python
numbers = [1, 2, 3]
try:
    print(",".join(numbers))
except TypeError as e:
    print("TypeError:", e)

print(",".join(str(n) for n in numbers))
````

### Replacing and counting

````python
s = "banana"
print(s.replace("a", "o"))
print(s.replace("a", "o", 2))       # only the first 2
print(s.count("a"), s.count("an"), s.count("z"))
````

### Searching: `find` and `index`
Both give the position of the first match. They differ when nothing is found.

* `find` returns `-1`
* `index` raises `ValueError`

````python
s = "banana"
print(s.find("n"), s.index("n"))
print(s.find("n", 3))               # start looking from position 3
print(s.find("z"))
try:
    s.index("z")
except ValueError as e:
    print("ValueError:", e)
````

**Edge case:** `find` returns `0` when the match is at the very start, and `0` is falsy. Do not write `if s.find("b"):`.

````python
s = "banana"
print(s.find("b"))
if s.find("b"):
    print("found")
else:
    print("this wrongly says not found")

if s.find("b") != -1:
    print("found (correct test)")
````

### Starts, ends and character tests

````python
name = "report_2024.txt"
print(name.startswith("report"), name.endswith(".csv"))
print(name.endswith((".txt", ".csv")))      # a tuple means "any of these"
print("123".isdigit(), "12a".isdigit(), "abc".isalpha(), "   ".isspace())
````

## 6. Formatting: f-strings and format specifications

An **f-string** starts with `f`. Put any expression inside `{ }`.

````python
name = "Asha"
marks = 87.456
print(f"{name} scored {marks}")
print(f"Double of marks is {marks * 2}")
print(f"{name!r}")              # !r shows the repr (with quotes)
print(f"{marks = }")            # handy for debugging
````

After a colon you can give a **format specification**: `{value:width.precisionType}`.

````python
x = 3.14159265
print(f"{x:.2f}")          # 2 digits after the point
print(f"{x:10.3f}|")       # width 10, right aligned (default for numbers)
print(f"{'ab':<6}|{'ab':>6}|{'ab':^6}|")    # left, right, centre
print(f"{'ab':*^8}")       # fill with *
print(f"{42:05d}")         # pad with zeros
print(f"{1234567:,}")      # thousands separator
print(f"{0.256:.1%}")      # percent
print(f"{255:b} {255:x} {255:o}")    # binary, hex, octal
````

A small table made with widths:

````python
rows = [("Asha", 91.5), ("Ravi", 78.5), ("Meera", 100)]
print(f"{'Name':<8}{'Marks':>8}")
for n, m in rows:
    print(f"{n:<8}{m:>8.1f}")
````

**Edge case:** to print a literal curly bracket in an f-string, double it. Also, use the other kind of quote inside the braces.

````python
v = 5
print(f"{{v}} is written as {{v}}, its value is {v}")
d = {"key": "value"}
print(f"{d['key']}")
````

## 7. Comparing strings and `in`

`==` compares text exactly. Case matters. `<` and `>` compare **dictionary order** using the code number of each character,
so all capital letters come before all small letters.

````python
print("abc" == "abc", "abc" == "ABC")
print("apple" < "banana", "Zebra" < "apple")      # capital Z is before small a
print("abc".lower() == "ABC".lower())              # case-insensitive comparison
print("10" < "9")                                  # text order, not number order
print(10 < 9)
````

`in` tests if a piece of text appears **anywhere** inside a string.

````python
print("th" in "Python", "x" in "Python", "" in "Python")
print("py" in "Python")             # case matters
print("Python".find("th") != -1)    # same as 'in', but 'in' is easier to read
````

## 8. Unicode and `len`

Python 3 strings hold **Unicode** characters, so Hindi, emoji and accented letters all work. `len` counts characters
(more exactly, code points), not bytes.

````python
words = ["cat", "café", "नमस्ते", "😀"]
for w in words:
    print(w, "| len:", len(w), "| bytes in UTF-8:", len(w.encode("utf-8")))
````

`ord` gives the code number of a character and `chr` does the opposite.

````python
print(ord("A"), ord("a"), chr(66), chr(2325))
print("\u0928")          # an escape written with a code number
````

**Edge case:** what you see as one character may be several code points, so `len` can surprise you.
The Hindi word above looks short but has 6 code points because vowel signs and joiners count separately.
Also, the letter `é` can be written as one code point or as `e` plus a combining accent.

````python
one = "\u00e9"            # e with accent, one code point
two = "e\u0301"           # e followed by a combining accent
print(one, two, len(one), len(two), one == two)
````

## 9. Text files

Programs keep data in **files** so it survives after the program ends. A **text file** holds characters.
The steps are: open the file, read or write, close the file.

The safe way is the `with` statement. It closes the file for you, even if an error happens.

```python
with open("notes.txt", "w", encoding="utf-8") as f:
    f.write("hello\n")
```

| Mode | Meaning |
|---|---|
| `"r"` | read (default). The file must exist. |
| `"w"` | write. Creates the file, **erases** old content. |
| `"a"` | append. Creates the file, adds to the end. |

Always give `encoding="utf-8"` for text files, so the same file reads the same way on every computer.

In this notebook all files are made inside a **temporary folder** that is deleted at the end.
Run the cells in order.

````python
import os, tempfile

tmp = tempfile.TemporaryDirectory()      # a throw-away folder
folder = tmp.name
path = os.path.join(folder, "notes.txt")
print("file path ends with:", os.path.basename(path))
````

### Writing
`write` does not add a new line for you. Add `\n` yourself.

````python
with open(path, "w", encoding="utf-8") as f:
    f.write("first line\n")
    f.write("second line\n")
    f.write("third line, with नमस्ते\n")

print("file closed:", f.closed)
print("size in bytes:", os.path.getsize(path))
````

### Reading the whole file

````python
with open(path, "r", encoding="utf-8") as f:
    content = f.read()
print(repr(content))
print(content)
````

### Reading line by line
* `readline()` reads one line each time it is called
* `readlines()` gives a list of all lines
* looping over the file object reads one line at a time, and is best for big files

Each line keeps its `\n` at the end.

````python
with open(path, encoding="utf-8") as f:
    print(repr(f.readline()))
    print(repr(f.readline()))
    print(repr(f.read()))       # whatever is left
    print(repr(f.readline()))   # at the end: empty string
````

````python
with open(path, encoding="utf-8") as f:
    print(f.readlines())

with open(path, encoding="utf-8") as f:
    for number, line in enumerate(f, start=1):
        print(number, line.rstrip("\n"))      # remove the newline we do not want to print twice
````

### Appending
Mode `"a"` adds to the end. Mode `"w"` would have erased the old lines.

````python
with open(path, "a", encoding="utf-8") as f:
    f.write("fourth line\n")

with open(path, encoding="utf-8") as f:
    print(len(f.readlines()), "lines now")

with open(path, "w", encoding="utf-8") as f:     # careful: this erases the file
    f.write("only line\n")
with open(path, encoding="utf-8") as f:
    print(f.read())
````

### Edge case: file not found
Reading a file that does not exist gives `FileNotFoundError`. Mode `"w"` and `"a"` create it.

````python
missing = os.path.join(folder, "nothing_here.txt")
print("exists:", os.path.exists(missing))
try:
    with open(missing, encoding="utf-8") as f:
        f.read()
except FileNotFoundError as e:
    print("FileNotFoundError:", e.strerror)

with open(missing, "a", encoding="utf-8") as f:    # append creates it
    pass
print("exists now:", os.path.exists(missing))
````

### Edge case: writing is not the same as printing
`write` accepts only strings, and it returns the number of characters written.

````python
with open(path, "w", encoding="utf-8") as f:
    n = f.write("abc\n")
    print("characters written:", n)
    try:
        f.write(42)
    except TypeError as e:
        print("TypeError:", e)
    f.write(str(42) + "\n")

with open(path, encoding="utf-8") as f:
    print(f.read())
````

### Edge case: newline handling
In text mode Python turns `\n` into the right line ending for your system when writing, and turns `\r\n` and `\r` back into `\n`
when reading. If you want to see the real bytes, open the file in binary mode `"rb"`.

````python
# Write a file using Windows line endings on purpose, as raw bytes
with open(path, "wb") as f:
    f.write(b"one\r\ntwo\r\n")

with open(path, "rb") as f:
    print("raw bytes :", f.read())
with open(path, encoding="utf-8") as f:
    print("text mode :", repr(f.read()))      # \r\n became \n
with open(path, encoding="utf-8", newline="") as f:
    print("newline='':", repr(f.read()))      # keeps \r\n as it is
````

Another trap: a file that does not end with a newline. Then the last line has no `\n`, so `rstrip("\n")` is safe but
`line[:-1]` would cut a real character.

````python
with open(path, "w", encoding="utf-8") as f:
    f.write("a\nb\nlast")

with open(path, encoding="utf-8") as f:
    print([line[:-1] for line in f])                    # wrong: 'last' became 'las'
with open(path, encoding="utf-8") as f:
    print([line.rstrip("\n") for line in f])            # right
````

Clean up the temporary folder when you are done.

````python
tmp.cleanup()
print("folder exists after cleanup:", os.path.exists(folder))
````

## 10. Quick summary

| Idea | Rule |
|---|---|
| Quotes | `'..'` and `".."` are the same, triple quotes span lines |
| Escapes | `\n`, `\t`, `\\`; raw string `r"..."` keeps backslashes |
| Index / slice | start at 0, negative counts from the end, stop is excluded |
| Out of range | index raises `IndexError`, slice is trimmed |
| Immutable | methods return new strings, the original never changes |
| `find` vs `index` | `find` gives `-1`, `index` raises `ValueError` |
| `split` / `join` | text to list, and list of strings back to text |
| f-string | `f"{value:>8.2f}"` |
| Compare | exact and case-sensitive, capitals come first |
| `len` | counts code points, not bytes |
| Files | `with open(path, mode, encoding="utf-8") as f` |
| Modes | `r` read, `w` write (erases), `a` append |

The `try` wraps `open`, so a missing file is turned into the answer `-1` instead of a crash.

```python
import os, tempfile

def count_lines_with(path, word):
    try:
        with open(path, encoding="utf-8") as f:
            return sum(1 for line in f if word in line)
    except FileNotFoundError:
        return -1

with tempfile.TemporaryDirectory() as folder:
    path = os.path.join(folder, "poem.txt")
    with open(path, "w", encoding="utf-8") as f:
        f.write("roses are red\nviolets are blue\nsugar is sweet\nare you?\n")
    print(count_lines_with(path, "are"))        # 3
    print(count_lines_with(os.path.join(folder, "none.txt"), "are"))   # -1
```
