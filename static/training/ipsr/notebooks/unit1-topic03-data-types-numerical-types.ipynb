## 2. What is a data type?

Every value in Python has a **type**. The type decides what the value is and what you can do with it.
You can add two numbers, but you cannot divide a word by a word.

Use `type()` to ask for the type of a value.

````python
print(type(42))
print(type(3.14))
print(type("hello"))
print(type(True))
print(type(None))
print(type(2 + 3j))
````

In Python the **value** carries the type, not the variable name. A name can point to an int now and a string later.

````python
x = 10
print(x, type(x))
x = "ten"
print(x, type(x))
````

## 3. The basic types at a glance

| Type | Meaning | Example |
|---|---|---|
| `int` | whole number, any size | `7`, `-3`, `0` |
| `float` | number with a decimal point | `3.14`, `-0.5`, `2e3` |
| `complex` | number with a real and imaginary part | `2 + 3j` |
| `bool` | truth value | `True`, `False` |
| `str` | text | `"hello"` |
| `NoneType` | "no value" | `None` |

This notebook focuses on the **numeric types** (`int`, `float`, `complex`, and `bool`, which behaves like a number).

## 4. Integers (`int`)

An `int` is a whole number, positive, negative or zero. There is **no size limit** except your computer's memory.

````python
a = 25
b = -7
print(a, b, type(a))
print(2 ** 100)            # ** means power; this number has 31 digits
````

You can write big numbers with underscores to make them easy to read. Python ignores the underscores.
You can also write integers in other bases: `0b` for binary, `0o` for octal, `0x` for hexadecimal.

````python
population = 1_400_000_000
print(population)

print(0b1010)     # binary 1010 = 10
print(0o17)       # octal 17 = 15
print(0xFF)       # hex FF = 255

print(bin(10), oct(15), hex(255))     # go the other way, to text
````

## 5. Floating point numbers (`float`)

A `float` is a number with a decimal point. Use `e` for powers of ten: `2.5e3` means 2.5 x 10^3 = 2500.

````python
pi = 3.14159
small = 1.5e-4      # 0.00015
big = 2.5e3         # 2500.0
print(pi, small, big)
print(type(big))    # still a float, even though it looks whole
print(5.0, type(5.0), 5 == 5.0)
````

Even a float with a whole value, like `5.0`, is a `float`. But `5 == 5.0` is `True`: they have equal value.

### Special float values
Floats can also be `inf` (infinity) and `nan` ("not a number").

````python
print(float("inf"), -float("inf"))
print(float("nan"))
print(1e308 * 10)       # too big for a float: becomes inf
````

## 6. Complex numbers (`complex`)

A complex number has a real part and an imaginary part. In Python the imaginary unit is written `j`, not `i`.
You will meet them in maths and engineering, so just see how they look.

````python
z = 3 + 4j
print(z, type(z))
print(z.real, z.imag)
print(abs(z))            # length of the number: sqrt(3*3 + 4*4)
print(z * z)             # (3+4j)*(3+4j) = 9 + 24j + 16 * (-1)
````

## 7. Booleans (`bool`)

A `bool` has only two values, `True` and `False` (capital first letter). They come from comparisons.

Surprise: `bool` is a **subclass of `int`**. `True` acts like 1 and `False` acts like 0.

````python
print(5 > 3, 5 == 6)
print(type(True))
print(True + True)            # 1 + 1
print(True * 10, False * 10)
print(isinstance(True, int))
marks = [True, False, True, True]
print("passed:", sum(marks))  # counting Trues is easy
````

## 8. `None`

`None` is a special value that means "nothing here". Its type is `NoneType`. It is not 0, not an empty string and not `False`.

````python
result = None
print(result, type(result))
print(result == 0, result == "", result == False)
print(result is None)        # the right way to test for None
````

## 9. Checking types: `type()` and `isinstance()`

`type(x) == int` checks for exactly that type. `isinstance(x, int)` also accepts subclasses and is the usual choice. You can give a tuple of types.

````python
x = 3.5
print(isinstance(x, float))
print(isinstance(x, (int, float)))       # int or float
print(isinstance(7, (int, float)))
print(type(True) == int, isinstance(True, int))    # the difference
````

## 10. Converting between types

Use the type name as a function: `int()`, `float()`, `str()`, `bool()`. This is called **type conversion** or **casting**.

````python
print(int("42") + 1)          # text to int
print(float("3.5") * 2)       # text to float
print(str(99) + " marks")     # number to text
print(int(7.9), int(-7.9))    # float to int cuts off the decimals (towards zero)
print(float(3))               # int to float
print(round(7.5), round(8.5), round(7.9))   # round is different from int
````

Python also converts automatically when you mix `int` and `float`: the result is a `float`. This is called **implicit conversion**.

````python
print(2 + 3.0, type(2 + 3.0))
print(True + 2.5)
````

## 11. Truthy and falsy values

When Python needs a yes/no answer (for example in `if`), it converts the value with `bool()`.
**Falsy** values: `0`, `0.0`, `0j`, `""` (empty text), `[]`, `{}`, `None`, `False`. Almost everything else is **truthy**.

````python
for v in [0, 1, -1, 0.0, 0.1, "", "0", "False", None, [], [0]]:
    print(f"{v!r:8} -> {bool(v)}")
````

## 12. Edge cases

### 12.1 Floats are not exact
A computer stores a float in binary. Many decimals, like 0.1, have no exact binary form, just as 1/3 has no exact decimal form.

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
print(f"{0.1:.20f}")        # show 20 digits: the stored value is slightly off
````

**Never compare floats with `==`.** Check that they are close, with `math.isclose`, or round first.

````python
import math
print(math.isclose(0.1 + 0.2, 0.3))
print(round(0.1 + 0.2, 2) == 0.3)
````

For money or exact decimals, use the `decimal` module, or count in paise (whole numbers).

````python
from decimal import Decimal
print(Decimal("0.1") + Decimal("0.2"))
print(Decimal("0.1") + Decimal("0.2") == Decimal("0.3"))
````

### 12.2 Integers are exact and huge, floats are not

````python
big = 10 ** 20
print(big + 1)                     # exact
print(float(big) + 1 == float(big))   # the +1 is lost in a float
print(2 ** 53 + 1, float(2 ** 53 + 1))   # float cannot hold this odd number
````

### 12.3 `nan` is not equal to anything, even itself

````python
nan = float("nan")
print(nan == nan)
print(nan != nan)
print(math.isnan(nan))       # the right test
````

### 12.4 Conversions that fail
`int("3.5")` fails, because the text is not a whole number. `int(3.5)` works. Go through `float` first.

````python
for text in ["42", "3.5", "abc", "", " 7 "]:
    try:
        print(repr(text), "->", int(text))
    except ValueError as e:
        print(repr(text), "-> ValueError:", e)

print(int(float("3.5")))
````

Notice that `" 7 "` works: `int()` ignores spaces at both ends.

### 12.5 Mixing types
Adding a number and text is an error. Convert one of them.

````python
try:
    print("Total: " + 5)
except TypeError as e:
    print("TypeError:", e)

print("Total: " + str(5))
print("Total:", 5)
````

### 12.6 `round` uses banker's rounding
When a number is exactly half way, Python's `round` goes to the **nearest even** number.

````python
print(round(0.5), round(1.5), round(2.5), round(3.5))
print(round(2.675, 2))     # 2.68 expected, but the stored value is slightly below 2.675
````

### 12.7 `bool("False")` is `True`
Any non-empty text is truthy, even the text `"False"`.

````python
print(bool("False"), bool(""))
````

## 13. Quick summary

| Type | Exact? | Size limit | Falsy value | Careful about |
|---|---|---|---|---|
| `int` | yes | none | `0` | `int(3.9)` cuts, it does not round |
| `float` | no (about 15 to 16 digits) | about 1.8e308 | `0.0` | `0.1 + 0.2 != 0.3`; use `math.isclose` |
| `complex` | no | like float | `0j` | imaginary unit is `j` |
| `bool` | yes | 2 values | `False` | `True + True == 2` |
| `str` | yes | memory | `""` | `bool("False")` is `True` |
| `NoneType` | yes | one value | `None` | test with `is None` |

Tools: `type(x)`, `isinstance(x, T)`, `int()`, `float()`, `str()`, `bool()`, `round()`.
