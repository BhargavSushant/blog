## 2. Quick reminder

A list is an ordered, **mutable** sequence. Most list methods change the list **in place** and return `None`.
Positions start at 0, and negative positions count from the end.

````python
fruits = ["apple", "banana", "cherry"]
print(fruits[0], fruits[-1], len(fruits))
````

## 3. Replacing items

Assign to an index to replace one item. Assign to a slice to replace several.

````python
marks = [70, 80, 90, 60]
marks[1] = 85               # replace one item
marks[-1] = 65              # negative index works too
print(marks)

marks[0:2] = [1, 2]         # replace a slice with the same number of items
print(marks)

marks[0:2] = [7, 8, 9]      # a slice can be replaced by a different number of items
print(marks)
````

**Edge case:** the index must already exist. You cannot create a new position by assigning past the end.

````python
marks = [70, 80]
try:
    marks[5] = 100
except IndexError as e:
    print("IndexError:", e)
````

## 4. Adding items: `append`, `insert`, `extend`

* `append(x)` adds **one** item at the end
* `insert(i, x)` adds one item **before** position `i`
* `extend(iterable)` adds **every item** of another list (or string, tuple ...) at the end

````python
nums = [1, 2, 3]
nums.append(4)
print(nums)

nums.insert(0, 0)           # at the front
print(nums)

nums.insert(2, 99)          # before position 2
print(nums)
````

### Edge cases for `insert`
`insert` never raises an error for the index. A position beyond the end acts like `append`. A negative position counts from the end,
and inserts **before** that item.

````python
a = [1, 2, 3]
a.insert(100, "end")        # too large: goes at the end
print(a)

b = [1, 2, 3]
b.insert(-1, "x")           # before the last item, NOT after it
print(b)

c = [1, 2, 3]
c.insert(-100, "start")     # too negative: goes at the front
print(c)
````

### `append` versus `extend`
This is a classic mix-up. `append` adds the whole object as **one** item. `extend` adds the items **inside** it.

````python
x = [1, 2]
x.append([3, 4])
print(x, len(x))

y = [1, 2]
y.extend([3, 4])
print(y, len(y))

z = [1, 2]
z += [3, 4]                 # same effect as extend
print(z)

w = []
w.extend("abc")             # a string is a sequence of characters
print(w)

v = [1, 2]
print(v + [3, 4], v)        # + makes a NEW list, v is unchanged
````

**Edge case:** `append` returns `None`. Do not write `lst = lst.append(x)`.

````python
lst = [1, 2]
lst = lst.append(3)
print(lst)                  # None: we lost the list
````

## 5. Removing items

| Tool | What it does | When it fails |
|---|---|---|
| `remove(x)` | removes the **first** item equal to `x` | `ValueError` if `x` is not in the list |
| `pop()` / `pop(i)` | removes and **returns** the item at `i` (default: last) | `IndexError` if the list is empty or `i` is too large |
| `del lst[i]` / `del lst[a:b]` | removes by position or slice, returns nothing | `IndexError` for a bad single index |
| `clear()` | removes everything | never |

### `remove`: by value, first match only

````python
letters = ["a", "b", "c", "b", "d"]
letters.remove("b")
print(letters)              # only the first "b" is gone

try:
    letters.remove("z")
except ValueError as e:
    print("ValueError:", e)

if "z" in letters:           # a safe way: test first
    letters.remove("z")
````

### `pop`: by position, and it gives the item back

````python
stack = [10, 20, 30, 40]
last = stack.pop()          # default: the last item
print(last, stack)

first = stack.pop(0)
print(first, stack)

try:
    [].pop()
except IndexError as e:
    print("IndexError:", e)

try:
    [1, 2].pop(5)
except IndexError as e:
    print("IndexError:", e)
````

### `del` and `clear`

````python
n = [0, 1, 2, 3, 4, 5, 6]
del n[0]                    # one item
print(n)
del n[1:3]                  # a slice
print(n)
del n[::2]                  # every second item
print(n)

n.clear()
print(n)

try:
    del n[0]
except IndexError as e:
    print("IndexError:", e)
````

`del lst` (without brackets) removes the **name** itself, not only the items. After that the name does not exist.

````python
tmp = [1, 2]
del tmp
try:
    print(tmp)
except NameError as e:
    print("NameError:", e)
````

## 6. Searching: `in`, `index`, `count`

````python
colours = ["red", "green", "blue", "green"]
print("blue" in colours, "pink" in colours)     # True / False
print("pink" not in colours)

print(colours.index("green"))        # position of the FIRST match
print(colours.index("green", 2))     # start searching at position 2
print(colours.count("green"), colours.count("pink"))
````

**Edge case:** `index` raises `ValueError` if the item is missing (unlike `str.find`, which returns `-1`).
There is no `find` for lists. Test with `in` first, or use `try/except`.

````python
colours = ["red", "green", "blue"]

def position(lst, item):
    return lst.index(item) if item in lst else -1

print(position(colours, "blue"), position(colours, "pink"))

try:
    colours.index("pink")
except ValueError as e:
    print("ValueError:", e)
````

`in` and `index` use `==`, so `1`, `1.0` and `True` all match each other.

````python
print(1 in [1.0], True in [1], [1, 2].index(2.0))
````

Searching a list checks items one by one, so it is slow for very long lists. That is fine for this course.

## 7. Sorting

### `sort()` changes the list, `sorted()` makes a new one

````python
nums = [5, 2, 9, 1]

new_list = sorted(nums)         # nums is untouched
print(new_list, nums)

nums.sort()                     # nums itself is changed
print(nums)

print(sorted([5, 2, 9, 1], reverse=True))
print(sorted("hello"))          # sorted accepts any iterable and always returns a list
print(sorted((3, 1, 2)))
````

**Edge case (the famous trap):** `sort()` returns `None`.

````python
nums = [3, 1, 2]
result = nums.sort()
print(result)                    # None
print(nums)                      # the list is sorted

nums = [3, 1, 2]
nums = nums.sort()               # wrong: nums is now None
print(nums)
````

### `reverse=True` and `reverse()`
`reverse=True` sorts from big to small. The method `reverse()` only flips the current order. It does not sort.

````python
a = [3, 1, 2]
a.sort(reverse=True)
print(a)

b = [3, 1, 2]
b.reverse()                      # flips, no sorting
print(b)

print(list(reversed([3, 1, 2])))     # reversed() gives an iterator: wrap it in list()
print([3, 1, 2][::-1])               # slice trick, new list
````

### `key=`: sort by a rule
`key` is a function that Python calls on each item. The items are then sorted by the values the function returns.

````python
words = ["banana", "kiwi", "apple", "fig"]
print(sorted(words))                 # alphabetical
print(sorted(words, key=len))        # by length
print(sorted(words, key=len, reverse=True))

names = ["bob", "Alice", "carol", "Dave"]
print(sorted(names))                       # capitals first
print(sorted(names, key=str.lower))        # ignore case

pairs = [("Asha", 82), ("Ravi", 67), ("Meera", 91)]
print(sorted(pairs, key=lambda p: p[1]))   # by the second item
````

`lambda p: p[1]` is a tiny unnamed function that returns the second item. You can also write a normal `def`.

### Sorting is stable
**Stable** means that items with equal keys keep their original order. This lets you sort by one rule and then by another.

````python
students = [("Asha", "B"), ("Ravi", "A"), ("Meera", "B"), ("Karan", "A")]
by_grade = sorted(students, key=lambda s: s[1])
print(by_grade)         # within each grade the original order is kept
````

**Edge case:** `sorted` with `reverse=True` is also stable. Equal items keep their original order, they are not flipped.

````python
data = [("a", 1), ("b", 1), ("c", 2)]
print(sorted(data, key=lambda t: t[1], reverse=True))
````

### Edge case: mixed types cannot be sorted
Python will not guess how to order a number against a string.

````python
mixed = [3, "apple", 1]
try:
    mixed.sort()
except TypeError as e:
    print("TypeError:", e)

print(sorted(["10", "9", "100"]))              # strings sort by text: '10' before '9'
print(sorted(["10", "9", "100"], key=int))     # convert with key to sort by number
print(sorted([3, 1.5, True, 0]))               # ints, floats and bools mix fine
````

Fix the data first (for example, convert everything to numbers), then sort.

## 8. Edge case: removing items while looping

Never change the length of a list while you loop over it. The loop uses positions, so after a removal items shift left and the
loop **skips** the next item.

````python
nums = [1, 2, 2, 3, 4, 4, 5]
for n in nums:
    if n % 2 == 0:
        nums.remove(n)
print(nums)          # wrong: some even numbers are still there
````

Three safe ways:

````python
nums = [1, 2, 2, 3, 4, 4, 5]

# 1. build a new list (best)
odds = [n for n in nums if n % 2 != 0]
print(odds)

# 2. loop over a copy, change the original
a = nums.copy()
for n in nums[:]:
    if n % 2 == 0:
        a.remove(n)
print(a)

# 3. loop backwards by index
b = nums.copy()
for i in range(len(b) - 1, -1, -1):
    if b[i] % 2 == 0:
        del b[i]
print(b)
````

## 9. Edge case: copy versus alias

`b = a` gives the same list a second name. Methods such as `sort` and `append` through one name show up in the other.

````python
a = [3, 1, 2]
b = a
b.sort()
print(a, b, a is b)

c = [3, 1, 2]
d = c.copy()            # also: list(c) or c[:]
d.sort()
print(c, d, c is d)

original = [3, 1, 2]
ordered = sorted(original)      # sorted() always returns a new list
print(original, ordered)
````

## 10. Edge case: `lst * n` and nested lists

`[0] * 3` repeats the **same object** three times. For numbers this is fine, because they are immutable.
For inner lists, all three positions point to **one** list.

````python
row = [0] * 3
print(row)

grid = [[0] * 3] * 2           # two references to the SAME inner list
grid[0][0] = 9
print(grid)                    # both rows changed
print(grid[0] is grid[1])
````

````python
grid = [[0] * 3 for _ in range(2)]     # a new inner list each time
grid[0][0] = 9
print(grid)
print(grid[0] is grid[1])
````

## 11. Quick summary

| Task | Tool | Notes |
|---|---|---|
| Replace | `lst[i] = x`, `lst[a:b] = [...]` | index must exist |
| Add one | `append(x)`, `insert(i, x)` | `insert` never raises for the index |
| Add many | `extend(it)`, `+=` | `append([..])` adds a nested list |
| Remove by value | `remove(x)` | first match, `ValueError` if missing |
| Remove by position | `pop(i)`, `del lst[i]` | `pop` returns the item, `IndexError` if bad |
| Remove all | `clear()` | |
| Search | `in`, `index(x)`, `count(x)` | `index` raises `ValueError` |
| Sort in place | `sort()` | returns `None` |
| Sort to new list | `sorted(it)` | works on any iterable |
| Options | `reverse=True`, `key=func` | sort is stable |
| Copy | `copy()`, `list(l)`, `l[:]` | `b = a` is an alias |
| Repeat | `[x] * n` | nested lists are shared, use a comprehension |
