## 1.4 Variables, assignments, immutable variables

*HTML lab: Unit 1 lab, lesson 2*

A variable is a name that points to a value. `b = a` makes a second name for the **same** value. Numbers, strings and tuples are *immutable*: they cannot be changed in place, so "changing" them really makes a new value. Lists are *mutable*.

````python
a = 5
b = a
a = a + 1
print(a, b)          # b is unchanged

s = "cat"
try:
    s[0] = "b"
except TypeError as e:
    print("Error:", e)
print("b" + s[1:])    # build a new string instead

p = [1, 2]
q = p                # same list, two names
q.append(3)
print(p, q)
````

## 1.5 Arithmetic operators and expressions

*HTML lab: Unit 1 lab, lesson 3*

Operators: `+ - * / // % **`. `//` is floor division, `%` is remainder. Precedence (high to low): `**`, then unary minus, then `* / // %`, then `+ -`. Use brackets when in doubt.

````python
print(7 / 2, 7 // 2, 7 % 2, 2 ** 5)
print(2 + 3 * 4)
print((2 + 3) * 4)
print(-2 ** 2)       # power binds tighter than minus
print((-2) ** 2)
print(-7 // 2, -7 % 2)   # floor goes toward minus infinity
````

## 1.9 Loops (for, while)

*HTML lab: Unit 1 lab, lesson 6*

`for` repeats over a sequence; `while` repeats until a condition fails. `break` leaves the loop, `continue` skips to the next round. A `while` loop must change something, or it never ends.

````python
total = 0
for i in range(1, 6):
    total += i
    print("i =", i, "running total =", total)

n = 27
steps = 0
while n != 1:
    n = n // 2 if n % 2 == 0 else 3 * n + 1
    steps += 1
print("Collatz steps for 27:", steps)

for i in range(10):
    if i == 3: continue
    if i == 6: break
    print(i, end=" ")
print()
````

## 1.10 Short-circuit (lazy) evaluation

*HTML lab: Unit 1 lab, lesson 5*

`A and B` stops at A if A is False. `A or B` stops at A if A is True. The right side is never evaluated, which is why it is safe to write `y != 0 and x / y > 2`.

````python
def noisy(label, value):
    print("  evaluating", label)
    return value
print("False and ...:"); r = noisy("A", False) and noisy("B", True); print(" ->", r)
print("True or ...:");  r = noisy("A", True) or noisy("B", False);   print(" ->", r)
x, y = 10, 0
print("safe check:", y != 0 and x / y > 2)
````

## 1.13 Basic list operators: replacing, inserting, removing, searching and sorting

*HTML lab: Unit 1 lab, lesson 8*

````python
L = [40, 10, 30, 20]
L[0] = 50                 # replace
L.insert(1, 99)           # insert at index
L.append(5)               # add at end
print(L)
L.remove(99)              # remove by value
last = L.pop()            # remove by position (last)
print(L, "popped", last)
print(30 in L, L.index(30))     # search
print(sorted(L), L)             # sorted() leaves L alone
L.sort(reverse=True)            # sort() changes L
print(L)
````

## 1.14 Dictionary literals, adding and removing keys, accessing and replacing values, traversing

*HTML lab: Unit 1 lab, lesson 8*

````python
d = {"CSE": 60, "ECE": 40}
d["ME"] = 30              # add a key
d["CSE"] = 65             # replace a value
del d["ECE"]              # remove a key
print(d)
print(d.get("EEE", 0))    # safe access with default
for k, v in d.items():
    print(k, "->", v)
print(list(d.keys()), sum(d.values()))
````
