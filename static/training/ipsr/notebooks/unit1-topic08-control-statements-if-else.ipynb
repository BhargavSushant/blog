## 2. The `if` statement

Normally Python runs lines from top to bottom. An `if` statement lets it **skip** some lines.

```
if condition:
    lines that run only when the condition is True
```

The condition is anything that gives `True` or `False` (see the previous topic). The colon `:` and the indentation are required.

````python
temperature = 35
if temperature > 30:
    print("It is hot.")
    print("Drink water.")
print("This line always runs.")
````

Change `temperature` to `20` and run again. The two indented lines are skipped.

## 3. `if` and `else`

`else` gives the action for when the condition is false. Exactly one of the two blocks runs.

````python
marks = 38
if marks >= 40:
    print("Pass")
else:
    print("Fail")
````

Another everyday example: even or odd. The operator `%` gives the **remainder** after division. A number is even when the remainder after dividing by 2 is 0.

````python
for number in [10, 7, 0, -4, -3]:
    if number % 2 == 0:
        print(number, "is even")
    else:
        print(number, "is odd")
````

## 4. `elif`: more than two choices

`elif` means "else if". Python checks the conditions **in order** and runs the **first** block whose condition is true. All the rest are skipped. `else` at the end is the catch-all and is optional.

````python
def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 40:
        return "D"
    else:
        return "F"

for m in [95, 90, 89, 75, 60, 59, 40, 39, 0]:
    print(m, grade(m))
````

Look at the boundaries: 90 gives A and 89 gives B. Because of the order, `elif marks >= 75` is only reached when `marks` is below 90, so we do not need to write `marks >= 75 and marks < 90`.

### A separate `if` is not the same as `elif`
Several plain `if` statements in a row are **all** checked. With `elif`, only one branch can run.

````python
score = 95
print("-- separate ifs --")
if score >= 90:
    print("A")
if score >= 75:
    print("B")        # also runs!
if score >= 60:
    print("C")        # also runs!

print("-- if / elif --")
if score >= 90:
    print("A")
elif score >= 75:
    print("B")
elif score >= 60:
    print("C")
````

## 5. Nested conditions

You can put an `if` inside another `if`. The inner one is only checked when the outer one is true. Each level is indented 4 more spaces.

````python
age = 20
has_licence = True

if age >= 18:
    if has_licence:
        print("You may drive.")
    else:
        print("Get a licence first.")
else:
    print("You are too young to drive.")
````

Deep nesting is hard to read. Often you can flatten it with `and`, or with `elif`.

````python
age = 20
has_licence = True

if age >= 18 and has_licence:
    print("You may drive.")
elif age >= 18:
    print("Get a licence first.")
else:
    print("You are too young to drive.")
````

## 6. Conditional expression (one-line if-else)

When you only need to **choose a value**, Python has a short form:

```
value_if_true if condition else value_if_false
```

Read it as a sentence: "give A if the condition holds, otherwise B."

````python
age = 16
status = "adult" if age >= 18 else "minor"
print(status)

n = -7
print("absolute value:", n if n >= 0 else -n)

for x in [3, 8, 5]:
    print(x, "is", "big" if x > 4 else "small")
````

Use it only for simple choices. If it gets long, write a normal `if` statement.

## 7. Example: leap year

A year is a leap year if:

* it is divisible by 4, **and**
* it is **not** divisible by 100, **unless** it is also divisible by 400.

So 2024 is a leap year, 1900 is not (divisible by 100 but not 400), and 2000 is (divisible by 400).

````python
def is_leap(year):
    if year % 400 == 0:
        return True
    elif year % 100 == 0:
        return False
    elif year % 4 == 0:
        return True
    else:
        return False

for y in [2024, 2023, 1900, 2000, 2100, 1996]:
    print(y, is_leap(y))
````

The same rule in one line using logical operators. Check that both versions agree.

````python
def is_leap2(year):
    return (year % 4 == 0 and year % 100 != 0) or year % 400 == 0

print(all(is_leap(y) == is_leap2(y) for y in range(1, 3001)))
````

## 8. Edge cases and common mistakes

**1. `=` instead of `==`.** One `=` stores a value. Two `==` compare. Python refuses `if x = 5:` as a syntax error. (We use `compile` so the notebook keeps running.)

````python
try:
    compile("x = 5\nif x = 5:\n    print('five')", "<test>", "exec")
except SyntaxError as e:
    print("SyntaxError:", e.msg)
````

**2. Wrong order of `elif` branches.** The first true branch wins, so put the **most specific** condition first.
Here the order is reversed, and nobody can ever get an A.

````python
def bad_grade(marks):
    if marks >= 40:
        return "D"        # catches everything from 40 up
    elif marks >= 60:
        return "C"        # never reached
    elif marks >= 90:
        return "A"        # never reached
    else:
        return "F"

print(bad_grade(95), bad_grade(65), bad_grade(45), bad_grade(10))
````

**3. Boundaries: `>` versus `>=`.** Decide whether the limit itself belongs to the group. Always test the values **exactly on** the boundary and one on each side.

````python
def pass_strict(m):
    return m > 40         # 40 fails

def pass_normal(m):
    return m >= 40        # 40 passes

for m in [39, 40, 41]:
    print(m, pass_strict(m), pass_normal(m))
````

**4. A gap in the conditions.** Using only `>` and `<` can leave a value that matches nothing.

````python
def size(x):
    if x < 10:
        return "small"
    elif x > 10:
        return "big"
    # x == 10 falls through: nothing is returned

print(size(5), size(10), size(20))     # size(10) gives None
````

**5. `else` belongs to the `if` at the same indentation (the dangling else).** Only the spaces decide which `if` an `else` goes with. These two pieces differ only in the indent of `else`, and they behave differently.

````python
x, y = 5, -1

if x > 0:
    if y > 0:
        print("both positive")
    else:                      # belongs to the INNER if
        print("x positive, y not positive")

if x > 0:
    if y > 0:
        print("both positive")
else:                          # belongs to the OUTER if
    print("x is not positive")
````

The first block prints a line. The second prints nothing, because `x > 0` is true so its `else` is skipped, and the inner `if` fails with no `else`.

**6. Forgetting the colon or the indent** gives `SyntaxError` / `IndentationError`. Always check both.

**7. Comparing text needs the exact case.** `"Yes" == "yes"` is `False`. Use `.lower()` first when you read answers from users.

````python
answer = "YES"
if answer == "yes":
    print("matched without lower()")
else:
    print("no match without lower()")

if answer.lower() == "yes":
    print("matched with lower()")
````

**8. Testing a float with `==`.** Use `math.isclose` or a range check, as you saw in the previous topic.

**9. Checking the same thing twice.** `if x > 5: ... elif x <= 5: ...` is better written as `if x > 5: ... else: ...`. The `else` is clearer and cannot leave a gap.

## 9. Quick summary

| Form | Use it for |
|---|---|
| `if cond:` | do something only when `cond` is true |
| `if / else` | choose one of two actions |
| `if / elif / else` | choose one of many; the first true branch wins |
| nested `if` | a second question that only matters after the first |
| `a if cond else b` | choose a **value** in one line |

* Colon after the condition, and indent the block by 4 spaces.
* Put the most specific condition first in an `elif` chain.
* Test boundary values: just below, exactly on, and just above.
* `==` compares, `=` assigns.

    else:
