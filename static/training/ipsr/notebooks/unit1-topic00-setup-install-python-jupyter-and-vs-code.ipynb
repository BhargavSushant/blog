## 2. The pieces and how they fit

| Piece | What it is | Why you need it |
|---|---|---|
| **Python** | the language and the program that runs your code (the *interpreter*) | everything else sits on top of it |
| **pip** | the package installer that comes with Python | installs libraries such as pandas |
| **Virtual environment** (venv) | a private folder with its own Python and its own libraries | keeps projects from breaking each other |


The last arrow is where most problems happen: the notebook can be attached to a *different* Python from the one

## 4. Part B: pip, the package installer

`pip` downloads libraries from the Python Package Index (PyPI). Always call it as `python -m pip`. This guarantees that
you use the pip that belongs to **that** Python.

```powershell
python -m pip --version
python -m pip install --upgrade pip
python -m pip install requests
python -m pip list
python -m pip show requests
python -m pip uninstall requests
```

The next cell shows the same idea from inside Python: it asks the current interpreter for its own pip.

````python
import subprocess, sys

result = subprocess.run([sys.executable, "-m", "pip", "--version"], capture_output=True, text=True)
print(result.stdout.strip() or result.stderr.strip())
````

### Installed packages from inside Python
`importlib.metadata` tells you whether a package is installed in the Python that is running this cell, and which version.

````python
from importlib.metadata import version, PackageNotFoundError

for name in ["pip", "ipykernel", "notebook", "jupyterlab", "pandas", "numpy", "matplotlib"]:
    try:
        print(f"{name:12} {version(name)}")
    except PackageNotFoundError:
        print(f"{name:12} not installed")
````

**Edge cases**

* `pip` and `python` can belong to different installations. That is why we type `python -m pip`.
* The package name on PyPI is not always the import name: you `pip install scikit-learn` but `import sklearn`.
* `pip install` needs internet access. Behind a proxy you may need `python -m pip install --proxy http://host:port package`.

## 5. Part C: virtual environments

### 5.1 Why

Project A needs `pandas 1.5`. Project B needs `pandas 3.0`. If both install into the one global Python, one project breaks.
A **virtual environment** is a private copy of Python plus its own `site-packages` folder, so every project gets its own
libraries. It is also easy to delete and rebuild.

### 5.2 Create one

Open a terminal **inside your project folder** (in VS Code: *Terminal, New Terminal*). Then:

```powershell
cd "D:\SAGE - IXP\ipsr"
python -m venv .venv
```

This makes a folder called `.venv`. The dot at the start is a convention that marks it as an "environment folder".

### 5.3 Activate it

| Shell | Command |
|---|---|
| Windows PowerShell | `.venv\Scripts\Activate.ps1` |
| Windows Command Prompt | `.venv\Scripts\activate.bat` |
| macOS / Linux | `source .venv/bin/activate` |

When it works, your prompt starts with `(.venv)`. To leave the environment, type `deactivate`.

### 5.4 Install into it

```powershell
python -m pip install --upgrade pip
python -m pip install pandas numpy matplotlib
python -m pip freeze > requirements.txt
```

`requirements.txt` lists exactly what you installed. A friend can rebuild your environment with
`python -m pip install -r requirements.txt`.

### 5.5 See it from Python

Inside a venv, `sys.prefix` points to the venv folder while `sys.base_prefix` points to the Python it was copied from.
They differ only inside a venv.

````python
import sys

def in_venv():
    return sys.prefix != sys.base_prefix

print("prefix      :", sys.prefix)
print("base_prefix :", sys.base_prefix)
print("Inside a virtual environment?", in_venv())
````

Now let us **create** a throw-away environment from Python, in a temporary folder, and look inside it. This is the same
thing `python -m venv` does.

````python
import venv, tempfile, pathlib, subprocess, sys

with tempfile.TemporaryDirectory() as tmp:
    env_dir = pathlib.Path(tmp) / "demo-env"
    venv.create(env_dir, with_pip=False)          # with_pip=False keeps this demo fast

    print("Top-level contents:", sorted(p.name for p in env_dir.iterdir()))

    scripts = "Scripts" if sys.platform == "win32" else "bin"
    env_python = env_dir / scripts / ("python.exe" if sys.platform == "win32" else "python")

    out = subprocess.run([str(env_python), "-c", "import sys; print(sys.prefix != sys.base_prefix)"],
                         capture_output=True, text=True)
    print("Is the new interpreter inside a venv?", out.stdout.strip())
````

You should see folders such as `Scripts` (or `bin` on macOS and Linux), `Lib` and a file `pyvenv.cfg`.
The environment folder is deleted when the cell ends.

### 5.6 Edge cases and common errors

| What you see | Cause | Fix |
|---|---|---|
| `Activate.ps1 cannot be loaded because running scripts is disabled on this system` | PowerShell blocks scripts by default | run once: `Set-ExecutionPolicy -Scope CurrentUser RemoteSigned`, then try again. Or use Command Prompt and `activate.bat` |
| `(.venv)` does not appear | you are in the wrong folder, or the activation command failed | `cd` into the project folder and run the command again; check the error text |
| `pip install` puts packages in the wrong place | the venv is not active | check the prompt, or call the venv's Python directly: `.venv\Scripts\python.exe -m pip install pandas` |
| venv stops working after you move or rename the folder | a venv stores absolute paths | delete `.venv` and create it again from `requirements.txt` |
| `No module named venv` on Ubuntu | the `venv` module is packaged separately | `sudo apt install python3-venv` |
| the venv is huge in a zip or Git repository | people commit it by mistake | never share `.venv`; share `requirements.txt` and add `.venv/` to `.gitignore` |

## 7. Part E: make your virtual environment appear as a kernel

A notebook runs on a **kernel**. If you start Jupyter *inside* an active venv, the notebook already uses that venv.
But if you want one Jupyter (or VS Code) to offer several environments, you **register** each environment as a kernel:

```powershell
# with the venv ACTIVE
python -m pip install ipykernel
python -m ipykernel install --user --name ipsr --display-name "Python (IPSR)"
```

* `--name` is a short internal name (no spaces).
* `--display-name` is what you see in the kernel menu.

Check and manage kernels:

```powershell
jupyter kernelspec list
jupyter kernelspec uninstall ipsr
```

In a notebook choose it from **Kernel, Change kernel, Python (IPSR)**.

### Prove which Python a notebook is using

````python
import sys
print(sys.executable)
````

If the path contains `.venv`, you are on the right kernel. If it points to a system Python, switch kernels.

### Installing packages from a notebook
Use the `%pip` magic, **not** `!pip`. `%pip` installs into the kernel that is running the notebook.
`!pip` may call a different Python, and then `import` fails even though the install "succeeded".

```python
%pip install requests
```

### Edge cases

* After installing a package, **restart the kernel** (*Kernel, Restart*) if the import still fails.
* If the notebook says `ModuleNotFoundError` but `pip list` in your terminal shows the package, the notebook is on a different Python. Print `sys.executable` and compare.
* The next cell shows what a missing package looks like, without stopping the notebook.

````python
try:
    import a_package_that_does_not_exist
except ModuleNotFoundError as e:
    print("ModuleNotFoundError:", e)
    print("Fix: %pip install <name> in a cell, then restart the kernel if needed.")
````

## 9. A self-check you can run any time

Run this cell **in the environment you plan to use** (right kernel selected). It prints a short report.

````python
import sys, platform
from importlib.metadata import version, PackageNotFoundError

def check_setup():
    problems = []
    py = sys.version_info
    print(f"Python {py.major}.{py.minor}.{py.micro} at {sys.executable}")
    if py < (3, 10):
        problems.append("Python is older than 3.10. Install a newer one.")
    in_venv = sys.prefix != sys.base_prefix
    print("Virtual environment:", "yes" if in_venv else "NO")
    if not in_venv:
        problems.append("Not running inside a virtual environment. Select your .venv kernel.")
    try:
        print("ipykernel:", version("ipykernel"))
    except PackageNotFoundError:
        problems.append("ipykernel is missing. Run: python -m pip install ipykernel")
    print()
    print("All good!" if not problems else "Things to fix:")
    for p in problems:
        print(" -", p)

check_setup()
````
