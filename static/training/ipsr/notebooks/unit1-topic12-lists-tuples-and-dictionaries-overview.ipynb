## 2. The big picture

A **container** holds several values under one name.

| | list | tuple | dict |
|---|---|---|---|
| Brackets | `[1, 2, 3]` | `(1, 2, 3)` | `{"a": 1, "b": 2}` |
| Ordered | yes | yes | yes (keeps insertion order) |
| Mutable (can change) | **yes** | **no** | **yes** |
| Items found by | position (0, 1, 2 ...) | position | **key** |
| Duplicates | allowed | allowed | keys must be unique |

**Mutable** means the object can be changed after it is created. **Immutable** means it cannot.

````python
marks = [72, 85, 60]                  # list
point = (3, 4)                        # tuple
student = {"name": "Asha", "age": 18} # dict (dictionary)

print(marks, type(marks))
print(point, type(point))
print(student, type(student))
````

## 3. Lists

A list is an ordered, changeable sequence. Items can be of different types.

````python
items = [10, "ten", 3.5, True, [1, 2]]
print(items)
print(len(items))
print(items[0], items[-1])      # first and last
print(items[1:3])               # slice gives a new list
print([])                       # empty list
print(list("abc"), list(range(4)))     # make a list from something else
````

Because a list is mutable, you can change it in place.

````python
marks = [72, 85, 60]
marks[0] = 75           # replace an item
marks.append(90)        # add at the end
print(marks)
````

## 4. Tuples

A tuple is an ordered sequence that **cannot be changed**. It is good for a fixed group of related values, such as a point
`(x, y)` or a date `(day, month, year)`.

````python
point = (3, 4)
print(point[0], point[-1], len(point))
print(point + (5,))             # joining makes a NEW tuple

try:
    point[0] = 10
except TypeError as e:
    print("TypeError:", e)
````

### Edge case: the one-element tuple
It is the **comma** that makes a tuple, not the brackets. `(1)` is just the number 1 in brackets.

````python
not_a_tuple = (1)
one = (1,)
also_one = 1,
print(type(not_a_tuple), type(one), type(also_one))
print(type(()))               # the empty tuple has no comma
print(len(one))
````

### Packing and unpacking
**Packing** puts values in a tuple. **Unpacking** takes them out into separate names.

````python
packed = 1, 2, 3                  # packing: brackets are optional
a, b, c = packed                  # unpacking: one name per item
print(a, b, c)

x, y = 10, 20
x, y = y, x                       # swap without a temporary variable
print(x, y)

first, *rest = [1, 2, 3, 4]       # * takes "the rest" as a list
print(first, rest)

try:
    p, q = (1, 2, 3)
except ValueError as e:
    print("ValueError:", e)
````

A function that returns "two values" really returns one tuple.

````python
def min_max(values):
    return min(values), max(values)

result = min_max([4, 9, 1])
print(result, type(result))
low, high = min_max([4, 9, 1])
print(low, high)
````

## 5. Dictionaries

A dictionary stores **key: value** pairs. You look up a value by its key, not by a position.

````python
student = {"name": "Asha", "age": 18, "city": "Bhopal"}
print(student["name"])
print(len(student))             # number of pairs
print(student.keys())
print(student.values())
print(student.items())
````

Add or change a pair by assigning to a key. Remove one with `del`.

````python
student["age"] = 19            # change
student["email"] = "a@x.com"   # add
del student["city"]            # remove
print(student)
````

**Order:** since Python 3.7 a dictionary remembers the order in which keys were added.

````python
d = {}
d["z"] = 1
d["a"] = 2
d["m"] = 3
print(d)
print(list(d))        # looping or list() gives keys in insertion order
````

### Keys must be hashable
Python finds a key quickly by using its **hash**, a number made from the key's value. So a key must never change.
Immutable things (numbers, strings, tuples of immutable things) are hashable. Lists and dictionaries are not.

````python
good = {1: "int", "a": "str", (1, 2): "tuple", 2.5: "float"}
print(good)

try:
    bad = {[1, 2]: "list"}
except TypeError as e:
    print("TypeError:", e)

print((1, 2) in good)
print(hash("abc") == hash("abc"))
````

**Edge case:** `1`, `1.0` and `True` are equal, so they count as the same key. The second assignment replaces the value
but keeps the first key.

````python
d = {1: "one"}
d[1.0] = "float one"
d[True] = "true"
print(d)
````

Values can be anything. Keys must be unique: a repeated key overwrites the earlier value.

````python
print({"a": 1, "a": 2})
````

### Missing keys and `.get`
`d[key]` raises `KeyError` when the key is absent. `d.get(key)` returns `None` instead, or a default you choose.

````python
student = {"name": "Asha", "age": 18}
try:
    print(student["phone"])
except KeyError as e:
    print("KeyError:", e)

print(student.get("phone"))
print(student.get("phone", "not given"))
print("phone" in student)
````

**Edge case:** `.get` with a default does not store anything. Use `setdefault` if you want the default saved too.

````python
d = {}
print(d.get("x", 0), d)              # d is still empty
print(d.setdefault("x", 0), d)       # now stored
````

## 6. Common operations on all three

| Task | list | tuple | dict |
|---|---|---|---|
| Number of items | `len(x)` | `len(x)` | `len(x)` (pairs) |
| Membership | `v in x` | `v in x` | `k in x` (**keys**) |
| Loop | `for v in x` | `for v in x` | `for k in x` |
| Index | `x[0]` | `x[0]` | `x[key]` |

````python
lst = [10, 20, 30]
tup = (10, 20, 30)
dct = {"a": 10, "b": 20, "c": 30}

print(len(lst), len(tup), len(dct))
print(20 in lst, 20 in tup, 20 in dct, "b" in dct)     # for a dict, "in" looks at keys
print(20 in dct.values())                               # to test values, ask for them
````

````python
for v in lst:
    print("list item", v)

for i, v in enumerate(tup):          # position and value together
    print("tuple item", i, v)

for k in dct:
    print("dict", k, dct[k])

for k, v in dct.items():             # key and value together (unpacking!)
    print(k, "->", v)
````

**Edge case:** a dictionary has no positions. `dct[0]` is a *key lookup* for the key `0`.

````python
dct = {"a": 10, "b": 20}
try:
    print(dct[0])
except KeyError as e:
    print("KeyError:", e)
print(list(dct)[0])            # first key, by turning the keys into a list
````

## 7. Nesting

Containers can hold other containers. This is how real data looks.

````python
students = [
    {"name": "Asha", "marks": [72, 85, 60]},
    {"name": "Ravi", "marks": [55, 40, 70]},
]
print(students[0]["name"])
print(students[1]["marks"][2])        # list -> dict -> list -> item

for s in students:
    print(s["name"], "average:", round(sum(s["marks"]) / len(s["marks"]), 1))

grid = [[1, 2, 3], [4, 5, 6]]          # a list of lists is a table
print(grid[1][0])
````

## 8. Edge case: a tuple that contains a list

A tuple is immutable, but this only means **the tuple cannot point to a different object**. If one of the objects inside is
mutable, that object can still change.

````python
t = (1, [10, 20], 3)
t[1].append(30)               # changes the list inside, the tuple itself is not reassigned
print(t)

try:
    t[1] = [99]               # trying to replace the item in the tuple
except TypeError as e:
    print("TypeError:", e)

try:
    t[1] += [40]              # += does both steps: the list changes, then the assignment fails
except TypeError as e:
    print("TypeError:", e)
print(t)                      # the list still grew
````

It also means a tuple holding a list is **not hashable**.

````python
print(hash((1, 2, 3)) is not None)
try:
    hash((1, [2]))
except TypeError as e:
    print("TypeError:", e)
````

## 9. Edge case: copy versus alias

`b = a` does **not** copy a list. It gives the same list a second name. This is an **alias**. A change through one name shows through the other.

````python
a = [1, 2, 3]
b = a                  # alias: same object
b.append(4)
print(a, b, a is b)

c = a.copy()           # a real (shallow) copy, a new list
c.append(5)
print(a, c, a is c)
````

`copy()` is **shallow**: the new container is separate, but the items inside are shared. For nested data this matters.

````python
import copy

grid = [[1, 2], [3, 4]]
shallow = grid.copy()
shallow[0].append(99)          # changes the inner list shared by both
print(grid)

deep = copy.deepcopy(grid)     # copies everything, inner lists too
deep[1].append(77)
print(grid)
print(deep)
````

The same is true for dictionaries (`d.copy()`). Tuples need no copy because they cannot change, but a tuple can hold a mutable item.

Why aliasing helps: a function that receives a list can change the caller's list.

````python
def add_bonus(marks):
    marks.append(5)        # changes the caller's list

m = [50, 60]
add_bonus(m)
print(m)
````

## 10. Which one should I use?

| Question | Use |
|---|---|
| A collection that will grow, shrink or be sorted | list |
| A fixed group of values (a point, a date, a return value) | tuple |
| Values you look up by name or ID | dict |
| A key in a dictionary made of several parts, such as `(row, col)` | tuple |
| A record with named fields (name, age, city) | dict |
| Many records of the same kind | list of dicts |

````python
# Which structure fits best? Examples:
temperatures = [31.5, 29.0, 33.2]                     # many readings, may grow: list
rgb_red = (255, 0, 0)                                 # fixed three values: tuple
phone_book = {"Asha": "98xxxx", "Ravi": "97xxxx"}     # look up by name: dict
distance = {("Bhopal", "Indore"): 190, ("Bhopal", "Delhi"): 740}   # tuple as key
print(distance[("Bhopal", "Indore")])
````

## 11. Quick summary

| Idea | Rule |
|---|---|
| list | ordered, mutable, `[ ]` |
| tuple | ordered, immutable, `( )`, one item needs a comma: `(1,)` |
| dict | key: value, keys unique and hashable, order of insertion kept, `{ }` |
| `in` | list and tuple test items, dict tests **keys** |
| `d[k]` vs `d.get(k)` | `KeyError` versus `None` or a default |
| Alias | `b = a` shares one object, `a.copy()` makes a new one |
| Shallow vs deep | `copy()` shares inner objects, `copy.deepcopy` does not |
| Tuple with a list | the list inside can still change |
| Empty values | `[]`, `()`, `{}` (`{}` is an empty dict) |
