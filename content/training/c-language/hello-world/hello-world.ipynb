## How a C program runs in this notebook

Colab runs Python by default, so we don't run C directly in a cell. Instead:

- `%%writefile hello.c` at the top of a cell saves the rest of the cell as a file named `hello.c`.
- A line starting with `!` runs a shell command. `!gcc -Wall -o hello hello.c` compiles the file into a program called `hello`, and `./hello` runs it.

This is the same thing you would do in a terminal on your own machine.

## 1. Your first program

```python
%%writefile hello.c
#include <stdio.h>

int main(void)
{
    printf("Hello, World!\n");
    return 0;
}
```

```python
!gcc -Wall -o hello hello.c && ./hello
```

What each line does:

| Line | Meaning |
|------|---------|
| `#include <stdio.h>` | Brings in the standard input/output library, which is where `printf` comes from. |
| `int main(void)` | Every C program starts running at `main`. It returns an `int` to the operating system. |
| `printf("...\n");` | Prints text. `\n` is a newline. Every statement ends with `;`. |
| `return 0;` | Tells the operating system the program finished without errors. |

The `&&` between the two commands means "run the program only if compiling succeeded".

## 2. Escape sequences and printing numbers

Inside a string, a backslash starts an **escape sequence**. `printf` also fills in values using **format specifiers** such as `%d` for whole numbers.

```python
%%writefile escapes.c
#include <stdio.h>

int main(void)
{
    printf("Name\tMarks\n");
    printf("Asha\t%d\n", 91);
    printf("Ravi\t%d\n", 78);
    printf("A quote: \"C is small\"\n");
    printf("A backslash: \\\n");
    return 0;
}
```

```python
!gcc -Wall -o escapes escapes.c && ./escapes
```

| Sequence | Prints |
|----------|--------|
| `\n` | newline |
| `\t` | tab |
| `\"` | a double quote |
| `\\` | a backslash |
| `%d` | the next `int` argument |

## 3. Reading a compiler error

Errors are normal. The skill is reading them. This program is missing a semicolon.

```python
%%writefile broken.c
#include <stdio.h>

int main(void)
{
    printf("This line has no semicolon\n")
    return 0;
}
```

```python
!gcc -Wall -o broken broken.c
```

Read the first line of the error: `broken.c:5:43` means **file `broken.c`, line 5, column 43**. GCC noticed the problem when it reached `return` on the next line, so it points at the end of line 5 and suggests where the `;` should go.

Fix the code in the cell above (add the `;`), run that cell again to rewrite the file, then run the compile cell again.
